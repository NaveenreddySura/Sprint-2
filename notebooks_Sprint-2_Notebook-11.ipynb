# Statistical Tests 
Statistical tests are used to analyze data, compare groups, identify relationships, and make decisions based on evidence. In business, they help us understand whether a change in sales, customer behavior, product performance, or marketing results is statistically meaningful.

### Assumptions
- The data is numerical.
- Observations are independent for independent-sample tests.
- Data is approximately normally distributed, especially for small samples.
- The standard independent t-test assumes equal variances; Welch's t-test does not.

### When to use
- To compare the average sales of two stores.
- To compare employee productivity before and after training.
- To compare the average spending of two customer groups.

### When not to use
- When the data is categorical.
- When comparing three or more independent group means; ANOVA may be more suitable.
- When the data contains paired observations but an independent-sample t-test is used.

### Business Example
An e-commerce company wants to check whether customers who receive promotional emails spend more money than customers who do not receive them. The company compares the average spending of both groups to determine whether the difference is statistically significant.

```python
from scipy.stats import ttest_ind

email_customers = [500, 600, 550, 700, 650]
regular_customers = [400, 450, 420, 480, 460]

t_stat, p_value = ttest_ind(
    email_customers,
    regular_customers,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference in average spending")
else:
    print("No significant difference in average spending")
```

### Interpretation
- P-value less than 0.05: There is statistically significant evidence that the average spending of the two groups differs.
- P-value greater than or equal to 0.05: There is insufficient evidence to conclude that the average spending differs.

## Z-Test

A Z-Test is a statistical test used to determine whether a sample mean or proportion differs significantly from a known or hypothesized population value.
A traditional one-sample Z-test for a mean assumes the population standard deviation is known.

### Assumptions
- The data is numerical for a test of means, or categorical for a test of proportions.
- Observations are independent.
- The population standard deviation is known for the traditional one-sample mean Z-test.
- The population is normal or the sample is sufficiently large for the normal approximation to be appropriate.
- Proportion tests require adequate expected counts for the normal approximation.

### When to use
- To compare a sample mean with a known population mean.
- To test whether a conversion rate differs from a target.
- To test whether a large-sample proportion has changed.

### When not to use
- When the population standard deviation is unknown and a small-sample mean test is needed; a t-test is generally more appropriate.
- When observations are dependent but the test assumes independence.
- When sample counts are too small for a proportion Z-test's normal approximation.

### Business Example
A company claims that its website conversion rate is 10%. After a redesign, the company checks 1,000 website visitors and finds that 120 made a purchase.
The company wants to test whether the conversion rate differs from 10%.

```python
from scipy.stats import norm

conversions = 120
visitors = 1000
expected_rate = 0.10

observed_rate = conversions / visitors

z_stat = (observed_rate - expected_rate) / (
    (expected_rate * (1 - expected_rate) / visitors) ** 0.5
)

p_value = 2 * norm.sf(abs(z_stat))

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Conversion rate differs significantly from 10%")
else:
    print("No significant difference from 10%")
```

###
Here, the observed conversion rate is 12%. The test checks whether that difference is statistically significant rather than simply due to sampling variation.

### Interpretation
- P-value less than 0.05: The observed conversion rate differs significantly from the claimed 10%.
- P-value greater than or equal to 0.05: There is insufficient evidence that the conversion rate differs from 10%.

## Chi-Square Test

A Chi-Square Test is a statistical test used to analyze categorical data by comparing observed frequencies with expected frequencies.
Two common types are:
- Chi-square goodness-of-fit test: Checks whether observed category counts follow an expected distribution.
- Chi-square test of independence: Checks whether two categorical variables are associated.

### Assumptions
- Data consists of counts or frequencies.
- Observations are independent.
- Categories are mutually exclusive.
- Expected cell counts are sufficiently large for the usual approximation to be reliable. A common guideline is that expected counts should generally be at least 5.

### When to use
- To determine whether gender and product preference are associated.
- To check whether customer preferences differ from an expected distribution.
- To analyze relationships between categorical variables.

### When not to use
- When the data consists of continuous measurements rather than counts.
- When observations are paired or repeated and the test does not account for that dependence.
- When expected counts are too small for the usual approximation; an exact test may be preferable.

### Business Example
A retail company wants to determine whether product preference is associated with customer membership status.
It collects the following data:
Customer type	Product A	Product B
Members	40	20
Non-members	25	35
The company wants to know whether product preference and membership status are associated.

```python
from scipy.stats import chi2_contingency

# Rows: members and non-members
# Columns: Product A and Product B
data = [
    [40, 20],
    [25, 35]
]

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:")
print(expected)

if p_value < 0.05:
    print("Product preference is associated with membership")
else:
    print("No significant association found")
```

### Interpretation
- P-value less than 0.05: There is evidence of an association between membership status and product preference.
- P-value greater than or equal to 0.05: There is insufficient evidence of an association.

## ANOVA (Analysis of Variance)

ANOVA is a statistical test used to determine whether the means of three or more groups differ significantly.
Instead of testing every pair separately, ANOVA initially tests whether all group means can reasonably be considered equal.

### Assumptions
- The dependent variable is numerical.
- Observations are independent.
- Data within each group is approximately normally distributed, particularly for small samples.
- Group variances are approximately equal for standard one-way ANOVA. Welch's ANOVA is an alternative when variances differ.

### When to use
- To compare average sales across three or more stores.
- To compare the effectiveness of multiple marketing campaigns.
- To compare employee productivity across several training methods.

### When not to use
- When the outcome is categorical rather than numerical.
- When observations are dependent and the analysis ignores the repeated-measures structure.
- When there are only two independent groups and a t-test is more straightforward.
- When assumptions are seriously violated and a more suitable alternative is available.

### Business Example
A company uses three marketing campaigns and wants to determine whether they produce different average sales.
- Campaign A: 200, 220, 210, 215
- Campaign B: 250, 260, 255, 245
- Campaign C: 300, 310, 290, 305

```python
from scipy.stats import f_oneway

campaign_a = [200, 220, 210, 215]
campaign_b = [250, 260, 255, 245]
campaign_c = [300, 310, 290, 305]

f_stat, p_value = f_oneway(
    campaign_a,
    campaign_b,
    campaign_c
)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("At least one campaign mean differs")
else:
    print("No significant difference among campaign means")
```

### Interpretation
- P-value less than 0.05: At least one campaign's mean sales differs significantly from another.
- P-value greater than or equal to 0.05: There is insufficient evidence of a difference among the group means.

### Mann-Whitney U Test

The Mann-Whitney U Test is a non-parametric statistical test used to compare two independent groups.
It is an alternative to the independent-samples t-test when the data is ordinal or when a rank-based comparison is more appropriate.
Instead of directly comparing means, it compares the ranks of observations across the two groups.

### Assumptions
- There are two independent groups.
- Observations are independent within and between groups.
- The outcome is at least ordinal, meaning values can be ranked.
- If the goal is to interpret the result as a difference in medians or typical values, the group distributions should have similar shapes.

### When to use
- When comparing two independent groups with highly skewed numerical data.
- When comparing customer satisfaction ratings.
- When comparing ordinal ratings, such as ratings from 1 to 5.
- When a rank-based alternative to the independent t-test is suitable.

### When not to use
- When comparing paired or repeated observations; the Wilcoxon signed-rank test may be appropriate.
- When comparing three or more independent groups without an appropriate extension.
- When the outcome is nominal categorical data with no meaningful ranking.
- When you specifically need to compare means; this test does not directly test the difference in means.

```python
from scipy.stats import mannwhitneyu

service_a = [2, 3, 3, 4, 5]
service_b = [4, 4, 5, 5, 5]

u_stat, p_value = mannwhitneyu(
    service_a,
    service_b,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("The groups differ significantly in their rank distributions")
else:
    print("No significant difference detected")
```

### 
In this example, Service B generally has higher ratings. The test evaluates whether the observed rank difference is statistically significant. It should not automatically be interpreted as a test of medians unless the distributions have similar shapes.

### Interpretation
- P-value less than 0.05: There is evidence that the two groups differ in their rank distributions.
- P-value greater than or equal to 0.05: There is insufficient evidence of a difference.

## Shapiro-Wilk Test

The Shapiro-Wilk Test is a statistical test used to determine whether a sample of numerical data is consistent with a normal distribution.
A normal distribution is a symmetric, bell-shaped distribution.
This test is commonly used to check the normality assumption before applying parametric tests such as the t-test or ANOVA.

### Assumptions
- The data is numerical.
- Observations are independent and representative of the population being studied.
- The sample contains enough observations for a meaningful assessment.
- The test is applied to the data whose normality matters, such as the residuals in a regression or ANOVA model.

### When to use
- Before applying a t-test or ANOVA, particularly with small samples.
- To assess whether a sample is consistent with a normal distribution.
- To help decide whether a parametric test is appropriate.

### When not to use
- When the data is categorical.
- When you want to compare two groups directly; Shapiro-Wilk does not compare group means.
- As the only method for checking normality, since plots and domain knowledge are also useful.
- As a strict decision rule for large datasets, where even minor deviations from normality can become statistically significant.

### Business Example
A company measures the time taken by employees to resolve customer support tickets.
Before comparing average resolution times between teams using a t-test, the analyst checks whether the data is consistent with a normal distribution.

```python

from scipy.stats import shapiro

resolution_times = [
    12, 14, 15, 13, 16,
    14, 15, 13, 17, 14
]

stat, p_value = shapiro(resolution_times)

print("Shapiro-Wilk statistic:", stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Evidence that the data is not normally distributed")
else:
    print("No significant evidence against normality")
```

### Interpretation
- P-value less than 0.05: Reject the null hypothesis of normality; the data shows evidence of deviation from a normal distribution.
- P-value greater than or equal to 0.05: Do not reject the null hypothesis; there is insufficient evidence to conclude that the data is non-normal.
