## What is Statistics?

Statistics is the process of collecting, organizing, analyzing, interpreting, and presenting data to understand information, identify patterns, and make better decisions based on the data.

```python
#Example:
customers = [20, 25, 30, 18, 27]

average = sum(customers) / len(customers)

print(average)
```

Here, the shop owner has collected the number of customers for five days. We use sum() to find the total number of customers and len() to find the number of days. Dividing the total by the number of days gives an average of 24 customers per day. This is an example of using statistics to understand business data.

## Why Statistics is Important in AI/ML?

Statistics is important in AI/ML because it helps us understand data, identify patterns, detect unusual values, measure uncertainty, and evaluate how well a model performs.

```python
#Example:
delivery_time = [25, 30, 35, 20, 40]

average = sum(delivery_time) / len(delivery_time)

print(average)
```

The company has collected delivery times for five orders and calculated an average delivery time of 30 minutes. Before building an ML model to predict delivery time, statistics can help the company understand things such as average delivery time, unusual delivery times, and patterns in the data.

## Statistics vs Machine Learning

Statistics mainly focuses on understanding and analyzing existing data, while Machine Learning uses data to learn patterns and make predictions or decisions about new data.

#Example:
Imagine a school has students' study hours and exam marks.

Study Hours → 2, 4, 6, 8
Marks       → 50, 60, 75, 90

Using statistics, we can analyze whether students who study more hours generally get higher marks. Using Machine Learning, we could train a model using this data and ask it to predict the marks of a student who studies for 7 hours. So, statistics helps us understand the relationship, while ML can learn the relationship and make a prediction.

## Descriptive Statistics
Descriptive statistics summarizes and describes the important characteristics of the data we already have using values such as average, minimum, maximum, median, and mode.

```python
#Example:
runs = [40, 55, 70, 30, 85]

print("Highest:", max(runs))
print("Lowest:", min(runs))
```

Here, we already have the player's scores from five matches. We use max() to find the highest score and min() to find the lowest score. We are simply describing the player's existing performance, so this is descriptive statistics.

## Inferential Statistics
Inferential statistics uses information collected from a sample to make an estimate or conclusion about a larger population.

#Example:
Population = 1,000,000 viewers
Sample = 1,000 viewers

850 people liked the movie

The company does not ask all one million viewers because that would be difficult and expensive. Instead, it asks 1,000 viewers. If 850 people like the movie, the company can use this sample to estimate the overall audience's opinion. This is inferential statistics.

## Population
Population is the complete group of people, objects, or data items that we are interested in studying.

#Example:
College students = 5,000

Population = All 5,000 students

The college is interested in the attendance of every student. Since all 5,000 students are part of the study, those 5,000 students together are called the population.

## Sample
A sample is a smaller group selected from a population to represent the larger population in a study.

#Example:
Population = 5,000 students
Sample = 300 students

Studying all 5,000 students may take more time. The college therefore selects 300 students and studies their attendance. These 300 students are the sample because they are only a smaller part of the complete population.

## Census
A census is a study in which information is collected from every member of the population instead of selecting only a sample.

#Example:
Employees = 2,000

Census = Salary information
         of all 2,000 employees

The company collects salary information from every employee instead of selecting only some employees. Since the complete population is included in the study, this is called a census.

## Sampling
Sampling is the process of selecting a smaller group of individuals or items from a larger population for statistical analysis.

```python
#Example
import random

customers = list(range(1, 10001))

sample = random.sample(customers, 100)

print(sample)
```

Here, we create 10,000 customer IDs to represent the population. random.sample() randomly selects 100 customers from those 10,000 customers. The selection process is called sampling, and the 100 selected customers form the sample.

## Variables
A variable is a characteristic or property of a person, object, or event that can have different values.

```python
#Example:
brand = "Toyota"
speed = 80
price = 1500000
```

Here, brand, speed, and price are variables. They store different types of information about the car. Another car could have a different brand, speed, or price, so these characteristics can have different values.

## Features
A feature is an input variable or measurable characteristic that is given to a Machine Learning model to help it make a prediction.

#Example:
customer = [25, 50000, 3]
25     → Customer age
50000  → Monthly income
3      → Number of previous purchases

The customer's age, income, and previous purchases provide useful information for predicting whether the customer will buy the product. These input values are called features. The ML model uses these features to make its prediction.

## Observation
An observation is one individual record or data point that represents one person, object, event, or item in a dataset.

#Example:
| City      | Temperature | Rain |
| --------- | ----------: | ---- |
| Bangalore |        28°C | No   |
| Chennai   |        32°C | Yes  |
| Delhi     |        25°C | No   |


Each row represents one observation. The first row contains information about Bangalore, the second row contains information about Chennai, and the third row contains information about Delhi. Therefore, this dataset contains three observations.

## Dataset
A dataset is a collection of related data containing multiple observations and variables that can be used for analysis or Machine Learning.

```python
#Example:
weather = [
    ["Bangalore", 28, "No"],
    ["Chennai", 32, "Yes"],
    ["Delhi", 25, "No"]
]

print(weather)
```

Here, we have weather information for three cities. Each city is one observation, while city name, temperature, and rain are variables. The complete collection of these records is called a dataset.

## Quantitative Data
Quantitative data is numerical data that represents quantities and can be measured or counted using numbers.

```python
#Example:
pushups = [15, 20, 25, 30, 18]

average = sum(pushups) / len(pushups)

print(average)
```

The values represent the number of push-ups performed by different people. Since the data consists of numbers representing a quantity that can be counted, it is quantitative data. We can also perform mathematical calculations such as finding the average.

## Qualitative data
Qualitative data is non-numerical data that describes categories, characteristics, qualities, or labels rather than measurable quantities.

```python
#Example:
food = ["Pizza", "Biryani", "Burger", "Pizza"]

print(food)
```

The values describe types of food rather than numerical quantities. We cannot calculate the average of Pizza and Biryani because they are categories. Therefore, food preference is qualitative data.

## Continuous Data
Continuous data is numerical data that can take any value within a range, including decimal values, and is usually obtained through measurement.

```python
#Example:
temperature = [36.5, 37.2, 38.1, 36.8]

print(temperature)
```

Body temperature is measured and can take decimal values such as 36.5°C, 36.55°C, or 36.57°C. Since temperature can take many possible values within a range, it is an example of continuous data.

## Discreate Data
Discrete data consists of separate, countable values, usually whole numbers, where values between two counts are not normally possible.

```python
#Example:
people = [50, 75, 100, 125]

print(people)
```

The values represent the number of people in the theatre. People can be counted as 50, 51, 52, and so on, but we cannot normally have 50.5 people. Since the values are countable whole numbers, this is discrete data.

## Structured Data
Structured data is data that follows a predefined format and is organized into clearly defined rows and columns, making it easy to store, search, and analyze.

#Example:
| Book ID | Book Name     | Author | Price |
| ------: | ------------- | ------ | ----: |
|     101 | Python Basics | John   |   500 |
|     102 | SQL Guide     | David  |   450 |
|     103 | AI Basics     | Alex   |   600 |


The library has predefined columns such as Book ID, Book Name, Author, and Price. Each book is stored in a separate row. Since the information follows a fixed structure of rows and columns, it is structured data.

## UnStructured Data
Unstructured data is information that does not follow a fixed format or predefined rows and columns, such as images, videos, audio, emails, and free-form text.

#Example:
"The delivery person was very helpful,
but my food arrived 20 minutes late."

This is a free-form customer message. It does not have fixed columns such as customer ID, delivery time, or rating. The customer can write anything in their own words. Therefore, this type of text is considered unstructured data.
