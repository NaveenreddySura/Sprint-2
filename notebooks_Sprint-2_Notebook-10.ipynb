# Hypothesis Testing

## Null Hypothesis
Definition

The Null Hypothesis, represented by \(H_0\), is the assumption that there is no significant difference, effect, or relationship in the population.

It is the hypothesis that we test against using statistical evidence.

### Formula

For a population mean:

$$ H_0:\mu=\mu_0 $$

where:

\(\mu\) = population mean
\(\mu_0\) = hypothesized mean

### Numerical Example

Suppose a company claims that the average delivery time is 30 minutes.

The null hypothesis is:

H

:μ=30

```python
from scipy.stats import ttest_1samp

delivery_times = [28, 31, 29, 32, 30, 27, 31, 29]

t_stat, p_value = ttest_1samp(delivery_times, 30)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

The Python code stores the sample delivery times and uses ttest_1samp() to compare the sample mean against the hypothesized population mean of 30 minutes. The function returns a test statistic and p-value, which we use to decide whether there is enough evidence to reject the null hypothesis.

## p-value
Definition

The p-value measures how unusual the observed data would be if the null hypothesis were true.

A small p-value provides stronger evidence against the null hypothesis.

### Formula

There is no single universal formula for every hypothesis test.

For a one-sample Z-test, for example:

$$ Z=\frac{\bar{x}-\mu_0}{\sigma/\sqrt{n}} $$

The p-value is then calculated from the probability distribution of the test statistic.

### Numerical Example

Suppose:

$$ p=0.03 $$

and:

$$ \alpha=0.05 $$

Since:

$$ 0.03<0.05 $$

we reject \(H_0\).

##AI/ML Engineer Example

An ML engineer comparing two model versions may use a statistical test to determine whether the observed improvement in model accuracy is likely to be a real difference rather than random variation.

```python
from scipy.stats import ttest_1samp

data = [52, 54, 51, 53, 55, 52, 54, 53]

t_stat, p_value = ttest_1samp(data, 50)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

The Python code performs a one-sample t-test comparing the sample against 50. It stores the resulting p-value and compares it with the significance level of 0.05. If the p-value is below 0.05, the code prints that we reject the null hypothesis.

### AI/ML Engineer Example

An ML engineer can calculate a confidence interval for model accuracy to understand the range in which the true performance is likely to lie.

```python
import numpy as np
from scipy.stats import t

data = [48, 50, 52, 49, 51, 50, 53, 47, 51, 49]

confidence_level = 0.95

mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

standard_error = std / np.sqrt(n)

t_critical = t.ppf(
    (1 + confidence_level) / 2,
    df=n - 1
)

margin_of_error = t_critical * standard_error

lower = mean - margin_of_error
upper = mean + margin_of_error

print("Mean:", mean)
print("95% Confidence Interval:", (lower, upper))
```

The Python code first calculates the sample mean and sample standard deviation. It then calculates the standard error and obtains the critical t-value using t.ppf(). The margin of error is calculated and added/subtracted from the mean to produce the lower and upper confidence limits.

## Significance Level
Definition

The Significance Level, represented by \(\alpha\), is the threshold used to decide whether evidence is strong enough to reject the null hypothesis.

A common significance level is 5%, represented as \(0.05\).

### Formula
α=0.05

### Numerical Example

Suppose:

$$ \alpha=0.05 $$

and:

$$ p=0.03 $$

Since:

$$ p<\alpha $$

we reject the null hypothesis.

### AI/ML Engineer Example

An ML engineer can choose \(\alpha=0.05\) when testing whether a new model's performance is significantly different from an existing model

```python
p_value = 0.03
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

The Python code stores the p-value as 0.03 and the significance level as 0.05. It compares the two values using an if condition. Because the p-value is smaller than alpha, the code concludes that there is enough statistical evidence to reject the null hypothesis.

## Type I Error
Definition

A Type I Error occurs when we reject the null hypothesis even though it is actually true.

It is also called a false positive.

### Formula

The probability of a Type I error is:

$$ P(\text{Type I Error})=\alpha $$
Numerical Example

If:

$$ \alpha=0.05 $$

then the test has a 5% significance level, meaning the procedure allows a 5% Type I error rate under the stated assumptions.

### AI/ML Engineer Example

Suppose an ML engineer concludes that a new model performs significantly better than the existing model when, in reality, there is no real improvement. This is a false positive.

```python
alpha = 0.05

print("Type I Error Rate:", alpha)
print("Type I Error Percentage:", alpha * 100, "%")
```

The Python code stores the significance level as 0.05. Since the probability of a Type I error is represented by alpha, the code displays both the decimal rate and percentage. At a 5% significance level, alpha corresponds to a 5% Type I error rate under the test's assumptions.

## Type II Error

A Type II Error occurs when we fail to reject the null hypothesis even though the alternative hypothesis is actually true.

It is also called a false negative.

## Formula

The probability of a Type II error is represented by:

$$ P(\text{Type II Error})=\beta $$

The statistical power of a test is:

Power=1−β

###Numerical Example

Suppose:

$$ \beta=0.20 $$

Then:

$$ \text{Power}=1-0.20=0.80 $$

So the test has:

$$ 80\% $$

power to detect the specified effect under the assumed conditions.

### AI/ML Engineer Example

An ML engineer concludes that a new model does not improve performance when it actually does. This is a false negative, representing a Type II error.

```python
beta = 0.20

power = 1 - beta

print("Type II Error Rate:", beta)
print("Power:", power)
print("Power Percentage:", power * 100, "%")
```

The Python code stores the Type II error probability as beta = 0.20. It then calculates statistical power using 1 - beta. Therefore, a Type II error probability of 20% corresponds to 80% power under the stated assumptions.

### Numerical Example

A company claims a new process increases productivity above 100 units per day.

We define:

$$ H_0:\mu\leq100 $$ $$ H_1:\mu>100 $$

If the sample provides sufficiently strong evidence that the mean is greater than 100, we reject \(H_0\).

### AI/ML Engineer Example

An ML engineer may use a one-tailed test when the question is specifically whether a new model's accuracy is greater than the existing model's accuracy.

```python
from scipy.stats import ttest_1samp

data = [105, 108, 110, 107, 112, 109, 106, 111]

t_stat, p_two_tailed = ttest_1samp(data, 100)

# Convert to a right-tailed p-value
if t_stat > 0:
    p_one_tailed = p_two_tailed / 2
else:
    p_one_tailed = 1 - (p_two_tailed / 2)

print("t-statistic:", t_stat)
print("One-tailed p-value:", p_one_tailed)
```

The Python code performs a one-sample t-test against 100. Since the test asks specifically whether the mean is greater than 100, we use a right-tailed test. The two-tailed p-value returned by ttest_1samp() is adjusted to obtain the right-tailed p-value when the test statistic is positive.

### AI/ML Engineer Example

An ML engineer can use a two-tailed test to determine whether the performance of a new model is different from the old model, without assuming beforehand that it will be better or worse.

```python
from scipy.stats import ttest_1samp

data = [32, 31, 29, 30, 33, 31, 28, 32]

t_stat, p_value = ttest_1samp(data, 30)

print("t-statistic:", t_stat)
print("Two-tailed p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

The Python code uses ttest_1samp() to compare the sample mean against 30. By default, the test is two-tailed, so it checks for differences in both directions. The resulting p-value is compared with 0.05 to decide whether there is sufficient evidence to reject the null hypothesis.
