## Data Distribution

## Normal Distribution

Normal Distribution is a continuous probability distribution where most values are concentrated around the mean and fewer values occur as we move away from the mean. It has a bell-shaped and symmetric curve.

### Formula
f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Where:

μ = Mean σ = Standard deviation x = Observation π = Mathematical constant

### Numerical Example

Suppose:

Mean = 50
Standard Deviation = 10

A value of 60 is:

60 - 50 = 10

So, it is 1 standard deviation above the mean.

### AI/ML Engineer Example

An ML engineer can use normal distribution to understand features such as measurement errors or model residuals and to detect values that are unusually far from the average.

```python
#Python Implementation

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

### Example

The heights of a large group of people can approximately follow a normal distribution.

### Explanation

Here, np.random.normal(50, 10, 1000) generates 1,000 random values with a mean of 50 and a standard deviation of 10. The histogram shows how these values are distributed, with most observations close to the mean.

## Uniform Distribution

Uniform Distribution is a distribution where all values within a specified range have an equal or approximately equal chance of occurring. It can be discrete or continuous.

### Formula

For continuous uniform distribution:
f(x) = 1 / (b - a)

for a ≤ x ≤ b.


where a is the minimum and b is the maximum value.

### AI/ML Engineer Example

An ML engineer can use uniform random values when initializing parameters, creating random samples, or generating synthetic data for testing an ML algorithm.

```python
#Python Implementation

import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(10, 20, 1000)

plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

Here, np.random.uniform(10, 20, 1000) generates 1,000 random values between 10 and 20. Since every part of the range has approximately equal probability, the histogram appears relatively flat.

## Gaussian Distribution

Gaussian Distribution is another name for the Normal Distribution. It is a bell-shaped, symmetric distribution where most observations are close to the mean and fewer observations occur farther away.

### Formula
f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

### Numerical Example

Suppose:

Mean = 100
Standard Deviation = 15

Then approximately:

68% of values → 85 to 115
95% of values → 70 to 130
99.7% of values → 55 to 145

### AI/ML Engineer Example

Gaussian distributions are commonly used in ML algorithms such as Gaussian Naive Bayes and are also useful for modeling normally distributed features.

```python
#Python Implementation

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(100, 15, 1000)

plt.hist(data, bins=30)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Gaussian Distribution")
plt.show()
```

Here, we generate 1,000 values with a mean of 100 and standard deviation of 15. The histogram shows that most values are close to 100, while fewer values appear farther away from the mean.

## Left Skew

A left-skewed distribution has a longer tail toward the left side of the distribution. Most observations are concentrated on the right side, with a few unusually low values pulling the distribution toward the left.

### Formula

There is no single formula required to identify left skew, but generally:

Mean<Median<Mode

### Numerical Example
Data = [10, 70, 75, 80, 82, 85, 90]

Here, the value 10 is much smaller than the other values, creating a longer tail toward the left.

### AI/ML Engineer Example

An ML engineer may encounter left-skewed features when most customers have high values but a small number have unusually low values. Recognizing skewness helps decide whether a transformation is needed before modeling.

```python
#Python Implementation

import matplotlib.pyplot as plt

data = [10, 70, 75, 80, 82, 85, 90]

plt.hist(data, bins=7)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Left Skewed Distribution")
plt.show()
```

Here, most values are between 70 and 90, while one value is only 10. This unusually low value creates a tail toward the left side, making the distribution left-skewed.

## Right Skew

A right-skewed distribution has a longer tail toward the right side of the distribution. Most observations are concentrated on the left side, with a few unusually high values pulling the distribution toward the right

### Formula

There is no single formula required to identify right skew, but generally:

Mode<Median<Mean

### Numerical Example
Data = [10, 12, 15, 18, 20, 22, 100]

Here, 100 is much larger than the other values, creating a long tail toward the right.

### AI/ML Engineer Example

An ML engineer may find right-skewed features such as customer spending, transaction amounts, or income. Transformations such as log() can sometimes help reduce strong right skew before modeling.

```python
#Python Implementation

import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 20, 22, 100]

plt.hist(data, bins=7)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Right Skewed Distribution")
plt.show()
```

Here, most values are between 10 and 22, but one value is 100. This unusually high value creates a long tail toward the right, making the distribution right-skewed.

## Symmetric Distribution

A symmetric distribution has a similar shape on both sides of its center. The left and right sides are approximately mirror images, and in a perfectly symmetric distribution, the mean and median are equal.

### Formula

For a perfectly symmetric distribution:

Mean=Median

### Numerical Example
Data = [10, 20, 30, 40, 50, 60, 70]

Mean:
Mean = 40

Median:
Median = 40

Therefore:

Mean = Median = 40

### AI/ML Engineer Example

An ML engineer checks whether numerical features are approximately symmetric to understand the shape of the data and decide whether preprocessing or transformations may be necessary.

```python
#Python Implementation

import statistics

data = [10, 20, 30, 40, 50, 60, 70]

mean = statistics.mean(data)
median = statistics.median(data)

print("Mean:", mean)
print("Median:", median)
```

Here, the values are evenly distributed around 40. The mean and median are both 40, showing that the data is centered symmetrically around the middle.

## Kurtosis

Kurtosis measures the shape of the tails of a distribution and helps indicate whether the dataset contains more or fewer extreme values compared with a normal distribution.

### AI/ML Engineer Example

An ML engineer can use kurtosis to identify features with extreme values or heavy tails, which may affect model performance and may require preprocessing.

```python
#Python Implementation

from scipy.stats import kurtosis

data = [10, 12, 14, 15, 16, 18, 20]

value = kurtosis(data)

print("Kurtosis:", value)
```

Here, kurtosis() calculates the kurtosis of the dataset. The result helps us understand the behavior of the distribution's tails. A higher kurtosis generally indicates a greater tendency for extreme values, while a lower kurtosis indicates lighter tails.

## Empirical Rule (68–95–99.7)

The Empirical Rule describes how data is distributed in a normal or approximately normal distribution. It states that about 68%, 95%, and 99.7% of observations fall within 1, 2, and 3 standard deviations from the mean.

### Formula

For a normal distribution:

68% of observations lie within 1 standard deviation of the mean. 95% of observations lie within 2 standard deviations of the mean. 99.7% of observations lie within 3 standard deviations of the mean. Formula / Rule Within 1 standard deviation:

μ ± 1σ → approximately 68%

Within 2 standard deviations:

μ ± 2σ → approximately 95%

Within 3 standard deviations:

μ ± 3σ → approximately 99.7%

### AI/ML Engineer Example

An ML engineer can use the Empirical Rule to understand normally distributed features and identify unusual observations. Values more than 3 standard deviations from the mean may be considered potential outliers.

```python
#Python Implementation

mean = 100
std = 15

print("68%:", mean - std, "to", mean + std)
print("95%:", mean - 2 * std, "to", mean + 2 * std)
print("99.7%:", mean - 3 * std, "to", mean + 3 * std)
```

Here, the mean delivery time is 100 minutes and the standard deviation is 15 minutes. According to the Empirical Rule, approximately 68% of deliveries will take between 85 and 115 minutes, 95% between 70 and 130 minutes, and 99.7% between 55 and 145 minutes.
