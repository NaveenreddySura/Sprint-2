# Probability Distributions
Probability distributions describe how probabilities are distributed across possible values or outcomes. They are very important in statistics, data science, and machine learning because they help us model uncertainty.

## Bernoulli Distribution

Bernoulli Distribution represents an experiment with only two possible outcomes: success or failure, usually represented as 1 or 0.
It is used for one single trial where the probability of success is p.

### Formula
P(X = x) = p^x (1-p)^(1-x)

Where:

p = probability of success 1-p = probability of failure x = 0 or 1

### Numerical Example
Suppose the probability that a customer purchases a product is 0.7.

Probability of Success = 0.7

Probability of Failure = 1 - 0.7 = 0.3

### Where It Is Used
- Yes/No experiments
- Customer purchase or no purchase
- Pass/Fail
- Spam/Not Spam
- Click/No Click
- Fraud/Not Fraud

### AI/ML Engineer Example
An ML engineer can use Bernoulli probability for binary classification, such as predicting whether a customer will buy a product: 1 = Buy, 0 = Not Buy.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)

plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.xticks([0, 1], ["Failure", "Success"])
plt.show()
```

In the Python code, p = 0.7 represents the 70% probability of success. We create two possible outcomes, 0 for failure and 1 for success, and calculate their probabilities as 0.3 and 0.7. The bar() function then displays these probabilities as two bars.

## Binomial Distribution

Binomial Distribution represents the number of successes in a fixed number of independent Bernoulli trials.
Each trial has only two outcomes: success or failure, with the same probability of success.

### Formula
P(X = k) = C(n,k) × p^k × (1-p)^(n-k)

Where:

n = number of trials k = number of successes p = probability of success

### Numerical Example
Suppose:

n = 5 p = 0.6 k = 3

We want the probability of exactly 3 successes.

P(X = 3) = C(5,3) × (0.6)^3 × (0.4)^2

P(X = 3) = 0.3456

Therefore, the probability is 34.56%.

### Where It Is Used
- Number of successful sales
- Number of defective products
- Number of customers who click an advertisement
- Number of successful predictions
- Number of heads in multiple coin tosses

### AI/ML Engineer Example
An ML engineer can use the binomial distribution to model the number of successful predictions or clicks among a fixed number of independent observations.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

In the Python code, n = 10 represents the number of trials and p = 0.5 represents the probability of success in each trial. np.arange() creates all possible numbers of successes from 0 to 10. binom.pmf() calculates the probability for each possible number of successes, and plt.bar() displays the resulting distribution.

## Poisson Distribution

Poisson Distribution describes the probability of a certain number of events occurring within a fixed time or space interval.
It is especially useful when events occur independently at a relatively constant average rate.

### Formula 
P(X = k) = (e^-λ × λ^k) / k!

Where:

λ = average number of events k = number of events e = Euler's numbe

### Numerical Example 
A customer support center receives an average of 3 calls per hour.

λ = 3

The probability of exactly 2 calls is:

P(X = 2) = (e^-3 × 3^2) / 2!

P(X = 2) ≈ 0.224

Therefore, the probability is approximately 22.4%.

### Where It Is Used
- Number of calls received per hour
- Website requests per minute
- Number of accidents per day
- Customer arrivals
- Server errors
- Defects per production unit

### AI/ML Engineer Example
An ML engineer can use Poisson distributions to model customer arrivals, server requests, failures, or support tickets per unit of time.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 3

x = np.arange(0, 12)
probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

plt.show()
```

In the Python code, lambda_value = 3 represents the average number of calls per hour. We create possible event counts from 0 to 11 using np.arange(). poisson.pmf() calculates the probability of each count, and the bar chart displays how the probabilities are distributed.

## Uniform Distribution

A Uniform Distribution is a distribution where values within a specified range have an equal probability of occurring.
For a continuous uniform distribution, every value within the interval has the same probability density.

### Formula 
For continuous Uniform Distribution:

f(x) = 1 / (b-a)

Where:

a = lower limit b = upper limit

### Where It Is Used
- Random number generation
- Simulation
- Random sampling
- Computer simulations
- Generating synthetic data
- Initializing some ML parameters

### AI/ML Engineer Example
Uniform distributions can be used when generating random samples or initializing parameters where every value within a particular range should have approximately equal probability.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(10, 20, 1000)

plt.hist(data, bins=20)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")

plt.show()
```

In the Python code, np.random.uniform(10, 20, 1000) generates 1,000 random values between 10 and 20. The hist() function groups these values into intervals and displays their frequency. Because the values are uniformly distributed, the histogram should look approximately flat.

## Normal Distribution

Normal Distribution is a continuous, bell-shaped distribution where most observations are close to the mean and fewer observations occur farther away.
It is symmetric around the mean, so the mean, median, and mode are equal in a perfectly normal distribution.

### Formula
f(x) = [1 / (σ√(2π))] × e^[-(x-μ)^2 / (2σ^2)]

Where:

μ = mean σ = standard deviation

### Numerical Example
Suppose:

Mean = 100 Standard Deviation = 10

According to the Empirical Rule:

68% → within 1 standard deviation 95% → within 2 standard deviations 99.7% → within 3 standard deviations

### Where It Is Used
- Heights
- Measurement errors
- Exam scores
- Manufacturing measurements
- Natural measurements
- Statistical modeling

### AI/ML Engineer Example
Normal distributions are commonly used to model measurement errors, model residuals, and naturally distributed features. Many statistical ML techniques also make assumptions related to normality.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(70, 10, 1000)

plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")

plt.show()
```

In the Python code, np.random.normal(70, 10, 1000) generates 1,000 values with a mean of 70 and standard deviation of 10. The histogram groups the generated values into ranges. The resulting graph should form an approximately bell-shaped curve, with most values concentrated around 70.

## Exponential Distribution

Exponential Distribution models the waiting time until the next event occurs when events happen independently at a constant average rate.
It is commonly used for time-related problems such as waiting time, service time, and time between events.

### Formula
f(x) = λe^(-λx), x ≥ 0

Where:

λ = event rate x = waiting time

### Numerical Example
Suppose a support center receives an average of 2 calls per hour.

λ = 2

The Exponential Distribution can model the waiting time until the next call.

### Where It Is Used
- Waiting time
- Time between customer arrivals
- Machine failure time
- Service time
- Time between website requests
- Queueing systems

### AI/ML Engineer Example
An ML engineer can use the exponential distribution to model time until the next server request, customer arrival, machine failure, or other event.

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_value = 2

data = np.random.exponential(
    scale=1 / lambda_value,
    size=1000
)

plt.hist(data, bins=30)

plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")

plt.show()
```

In the Python code, lambda_value = 2 represents an average event rate of 2 events per hour. The scale parameter is set to 1 / lambda_value, because the mean waiting time of an exponential distribution is 1/λ. np.random.exponential() then generates 1,000 waiting times, and the histogram shows their distribution.
