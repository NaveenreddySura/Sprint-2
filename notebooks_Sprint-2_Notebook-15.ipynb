## 1. Explain the difference between Mean and Median.
### Mean:
The mean is the average of all values in a dataset. It is calculated by adding all the values and dividing the total by the number of values.
### Median:
    The median is the middle value when the data is arranged in ascending or descending order. If there are an even number of values, the median is the average of the two middle values.
### Example:
Data: [10, 20, 30, 40, 100]
- Mean = 200 / 5 = 40
- Median = 30
The mean is affected by extreme values, whereas the median is more resistant to outliers.

## 2 What is Variance?
Variance is a statistical measure that describes how far the data values are spread out from their mean. A low variance indicates that values are close to the mean, while a high variance indicates that they are more widely spread.
### Population variance formula:
\[
\sigma^2=\frac{\sum(x_i-\mu)^2}{N}
\]

### Example:
Data: [2, 4, 6]
Mean = 4
### Population variance:
\[
\sigma^2=\frac{(2-4)^2+(4-4)^2+(6-4)^2}{3}
=\frac{8}{3}\approx2.67
\]

## 3 What is Standard Deviation?
Standard Deviation measures the spread of values around the mean and is calculated as the square root of the variance. Unlike variance, standard deviation is expressed in the same units as the original data, making it easier to interpret.
#### Population standard deviation formula:
\[
\sigma=\sqrt{\frac{\sum(x_i-\mu)^2}{N}}
\]

#### Example:
For [2, 4, 6], the population variance is approximately 2.67.
\[
\sigma=\sqrt{2.67}\approx1.63
\]

This means the standard deviation is approximately 1.63.

## 4 Explain Bayes' Theorem with an example.
Bayes' Theorem is a statistical formula used to calculate the probability of an event based on previously known information and new evidence. It updates the probability of an event when additional evidence becomes available.
Formula:
\[
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
\]

#### Example:
Suppose 40% of emails are spam. A particular word appears in 80% of spam emails and 10% of non-spam emails.
- \(P(Spam)=0.40\)
- \(P(Word|Spam)=0.80\)
- \(P(Word|NotSpam)=0.10\)
The probability that an email is spam given that it contains the word is:
\[
P(Spam|Word)=
\frac{0.80\times0.40}
{(0.80\times0.40)+(0.10\times0.60)}
\]

\[
=\frac{0.32}{0.38}\approx0.842
\]

Therefore, the probability is approximately 84.2%.

## 6. Explain Correlation and Covariance.
#### Correlation
measures the strength and direction of the linear relationship between two variables. Its value ranges from -1 to +1, making it useful for comparing relationships across different datasets.
#### Covariance 
measures how two variables change together. Positive covariance indicates that the variables tend to move in the same direction, while negative covariance indicates that they tend to move in opposite directions.

```python
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

mean_x = sum(x) / len(x)
mean_y = sum(y) / len(y)

covariance = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(len(x))) / len(x)

std_x = (sum((value - mean_x) ** 2 for value in x) / len(x)) ** 0.5
std_y = (sum((value - mean_y) ** 2 for value in y) / len(y)) ** 0.5

correlation = covariance / (std_x * std_y)

print("Covariance:", covariance)
print("Correlation:", correlation)
```

## 7. What is Skewness?
Skewness measures the asymmetry of a data distribution around its center. It helps us understand whether the data has a longer tail on the left or right side.
- Zero skewness: The distribution is approximately symmetric.
- Positive skewness: The right tail is longer; a few large values may pull the mean upward.
- Negative skewness: The left tail is longer; a few small values may pull the mean downward.
####Example:
Income data often has positive skewness because a small number of people earn much more than the majority.

## 8. Explain IQR.
Interquartile Range (IQR) measures the spread of the middle 50% of a dataset. It is calculated as the difference between the third quartile (Q3) and the first quartile (Q1), and it is commonly used to detect potential outliers.
Formula:
\[
IQR=Q3-Q1
\]

Outlier boundaries:
\[
Lower=Q1-1.5(IQR)
\]

\[
Upper=Q3+1.5(IQR)
\]

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

data.sort()

n = len(data)
mid = n // 2

lower_half = data[:mid]
upper_half = data[mid:]

q1 = (lower_half[1] + lower_half[2]) / 2
q3 = (upper_half[1] + upper_half[2]) / 2

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

## 9. What is a p-value?
A p-value is the probability, assuming the null hypothesis and the statistical test's assumptions are true, of obtaining a result at least as extreme as the one observed. It helps us assess the evidence against the null hypothesis.
Example: Suppose a hypothesis test produces a p-value of 0.03 and the significance level is 0.05.
Since \(0.03<0.05\), we reject the null hypothesis.
A p-value does not tell us the probability that the null hypothesis is true.

```python
data = [52, 48, 51, 49, 50, 53, 47, 51, 49, 50]

hypothesized_mean = 50
import math

data = [52, 48, 51, 49, 50, 53, 47, 51, 49, 50]
hypothesized_mean = 50

n = len(data)
mean = sum(data) / n

variance = sum((x - mean) ** 2 for x in data) / (n - 1)
std_dev = math.sqrt(variance)

t_value = (mean - hypothesized_mean) / (std_dev / math.sqrt(n))

print("Mean:", mean)
print("T-value:", t_value)
```

## 11. Write Python code to calculate Variance.

```python
data = [10, 20, 30, 40, 50]

mean_value = sum(data) / len(data)

population_variance = sum(
    (x - mean_value) ** 2 for x in data
) / len(data)

sample_variance = sum(
    (x - mean_value) ** 2 for x in data
) / (len(data) - 1)

print("Population Variance:", population_variance)
print("Sample Variance:", sample_variance)
```

We calculate the mean, find each value's squared difference from the mean, and add those squared differences. Dividing by the number of observations gives the population variance, while dividing by one less than the sample size gives the sample variance.

## 13. Calculate Correlation between two variables.

```python
import pandas as pd

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

df = pd.DataFrame({
    "X": x,
    "Y": y
})

correlation = df["X"].corr(df["Y"])

print("Correlation:", correlation)
```

We created a DataFrame containing two numerical variables and used Pandas' corr() method to calculate their Pearson correlation coefficient. The result is 1.0, indicating a perfect positive linear relationship because Y increases proportionally as X increases.
