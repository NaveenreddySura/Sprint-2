Statistics plays an important role in machine learning because it helps us understand data, handle missing values, detect outliers, select useful features, train models, and evaluate predictions. It provides mathematical methods to identify patterns in data and make predictions based on evidence.

## Data Cleaning

Data Cleaning is the process of identifying and correcting inaccurate, inconsistent, duplicated, or invalid values in a dataset before using it for analysis or machine learning. Statistics helps us understand the data's distribution, identify unusual values, and determine whether certain records require correction or further investigation.
### Formula
There is no single formula for data cleaning. Statistical summaries commonly used include the mean, median, standard deviation, and minimum and maximum values.
#### Numerical Example
Suppose a dataset contains these employee ages:
[22, 24, 25, 23, 240]
The value 240 is suspicious because it is far outside the expected range for human age. Statistical summaries and domain knowledge can help identify it for investigation.
#### Where It Is Used
- Identifying invalid values and duplicate records.
- Finding inconsistent data formats.
- Detecting suspicious numerical values.
- Preparing reliable data for machine learning.
#### AI/ML Engineer Example
An AI/ML engineer is building a salary-prediction model and finds that some employee ages are recorded as 240 instead of 24. The engineer investigates the suspicious records and corrects them only when the original values can be verified.

```python
import pandas as pd

data = {
    "Employee": ["A", "B", "C", "D", "E"],
    "Age": [22, 24, 25, 23, 240]
}

df = pd.DataFrame(data)

print("Before cleaning:")
print(df)

print("\nAge summary:")
print(df["Age"].describe())

# Flag suspicious ages for investigation
df["Suspicious_Age"] = (
    (df["Age"] < 18) | (df["Age"] > 100)
)

print("\nAfter checking:")
```

## Missing Value Handling

Missing Value Handling is the process of identifying and treating values that are absent from a dataset. Statistical methods help us decide whether missing values should be filled using the mean, median, mode, or a more advanced method, or whether certain rows or columns should be removed.

#### Numerical Example
Suppose employee salaries are:
[20000, 25000, 30000, missing]
The mean of the available salaries is:
\[
\bar{x}=\frac{20000+25000+30000}{3}=25000
\]

We can use 25,000 to fill the missing value as a simple baseline method.
#### Where It Is Used
- Handling missing customer information.
- Filling gaps in salary, sales, or measurement data.
- Preparing complete datasets for algorithms that cannot handle missing values directly.
#### AI/ML Engineer Example
An AI/ML engineer develops a model to predict customer spending. Some customers have missing income values, so the engineer uses median imputation to fill missing values while reducing sensitivity to unusually high incomes.

```python
import pandas as pd

data = {
    "Customer": ["A", "B", "C", "D"],
    "Income": [20000, 25000, None, 40000]
}

df = pd.DataFrame(data)

print("Before handling missing values:")
print(df)

median_income = df["Income"].median()

df["Income"] = df["Income"].fillna(median_income)

print("\nAfter handling missing values:")
print(df)
```

We created a DataFrame containing customer income values, with None representing a missing income. We used median() to calculate the median of the available incomes, which is 25,000 in this example. We then used fillna() to replace the missing value with that median. This produces a complete income column, although in real machine learning workflows the median should be calculated from the training data only to avoid data leakage.

## Outlier Detection
Definition
Outlier Detection is the process of identifying observations that differ substantially from the general pattern of a dataset. Statistics provides methods such as the Interquartile Range (IQR) and Z-score to detect unusual values that may represent errors, rare events, or meaningful exceptions.
#### Formula
IQR method:
\[
IQR=Q_3-Q_1
\]

Lower and upper bounds:
\[
L=Q_1-1.5(IQR)
\]

\[
U=Q_3+1.5(IQR)
\]

Values outside these bounds are flagged as potential outliers.
#### Numerical Example
Consider these values:
[10, 12, 13, 14, 15, 16, 18, 100]
Most values are between 10 and 18, but 100 is much larger than the others. The IQR method can identify 100 as a potential outlier.
#### Where It Is Used
- Detecting unusual customer transactions.
- Finding data-entry errors.
- Identifying abnormal sensor readings.
- Investigating unusually high or low sales.
#### AI/ML Engineer Example
An AI/ML engineer develops a fraud-detection model and identifies transactions with unusually large amounts. Rather than automatically deleting these transactions, the engineer investigates them because they could represent fraud or legitimate high-value purchases.

```python
import pandas as pd

data = [10, 12, 13, 14, 15, 16, 18, 100]

df = pd.DataFrame({"Value": data})

q1 = df["Value"].quantile(0.25)
q3 = df["Value"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[
    (df["Value"] < lower_bound) |
    (df["Value"] > upper_bound)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Potential outliers:")
print(outliers)
```

We created a DataFrame containing eight numerical observations and calculated the first quartile (q1) and third quartile (q3). We subtracted the quartiles to obtain the IQR and used it to calculate the lower and upper outlier bounds. Finally, we filtered the DataFrame to display values outside those bounds. The result identifies 100 as a potential outlier, which should be investigated before deciding whether to retain, transform, or remove it.

## Feature Engineering

Feature Engineering is the process of creating new input variables or transforming existing variables to make useful patterns easier for a machine learning model to learn. Statistics helps engineers discover meaningful combinations, ratios, aggregations, and transformations that may improve predictive performance.
#### Formula
A simple example is creating a total-spending feature:
\[
\text{Total Spending}=\text{Quantity}\times\text{Unit Price}
\]

#### Numerical Example
A customer purchases 4 products, each costing ₹500.
\[
\text{Total Spending}=4\times500=₹2000
\]

Instead of using only quantity and unit price, we can create a new feature called Total_Spending.
#### Where It Is Used
- Creating total-sales and average-spending features.
- Extracting month, day, or hour from dates.
- Creating ratios and interaction features.
- Improving the information available to a model.
#### AI/ML Engineer Example
An AI/ML engineer is building a customer-purchase prediction model. The engineer creates features such as total spending, average order value, and days since the last purchase to help the model understand customer behavior.

```python
import pandas as pd

data = {
    "Quantity": [2, 4, 3, 5],
    "Unit_Price": [100, 500, 200, 300]
}

df = pd.DataFrame(data)

df["Total_Spending"] = (
    df["Quantity"] * df["Unit_Price"]
)

print(df)
```

We created a DataFrame containing the quantity purchased and the unit price for four transactions. We then multiplied these two columns to create a new feature named Total_Spending. This derived feature directly represents the total value of each transaction and can provide a model with useful information about customer purchases.

## Feature Scaling

Feature Scaling is the process of adjusting numerical features to comparable scales so that variables with large numerical values do not unnecessarily dominate certain machine learning algorithms. Statistical measures such as the mean and standard deviation are used in standardization, while minimum and maximum values are used in min-max scaling.
#### Formula
Standardization (Z-score scaling):
\[
z=\frac{x-\mu}{\sigma}
\]

Where \(x\) is the original value, \(\mu\) is the feature mean, and \(\sigma\) is its standard deviation.
#### Numerical Example
Suppose a student's score is 80, the mean score is 60, and the standard deviation is 10.
\[
z=\frac{80-60}{10}=2
\]

The standardized score is 2, meaning the original score is two standard deviations above the mean.
### Where It Is Used
- Preparing features for K-Nearest Neighbors (KNN).
- Scaling inputs for Support Vector Machines (SVM).
- Improving optimization behavior in many gradient-based models.
- Preventing large-unit features from dominating distance calculations.
#### 
AI/ML Engineer Example
An AI/ML engineer builds a customer-segmentation model using age and annual income. Since income may be measured in tens of thousands while age is measured in tens, the engineer scales the features so that income does not dominate distance calculations merely because of its units.

```python
import pandas as pd

data = {
    "Age": [20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)

mean_age = df["Age"].mean()
std_age = df["Age"].std()

df["Age_Standardized"] = (
    (df["Age"] - mean_age) / std_age
)

print(df.round(2))
```

We created a DataFrame containing five ages and calculated their mean and sample standard deviation using mean() and std(). We then subtracted the mean from each age and divided the result by the standard deviation. This created a standardized feature with a mean of approximately zero. Values above the mean have positive standardized scores, while values below the mean have negative scores.

## Data Normalization

Data Normalization commonly refers to transforming numerical values to a fixed range, often between 0 and 1, using the feature's minimum and maximum values. In machine learning, this can help make features comparable, especially for distance-based algorithms; however, the term normalization can also refer to other transformations depending on the context.

#### Numerical Example
Suppose the minimum age is 20, the maximum age is 60, and the current age is 40.
\[
x'=\frac{40-20}{60-20}=0.5
\]

The normalized age is 0.5.
#### Where It Is Used
- Preparing data for KNN and clustering.
- Scaling image pixel values to the range 0–1.
- Making numerical features comparable.
- Preparing inputs for certain neural networks.
#### AI/ML Engineer Example
An AI/ML engineer builds a recommendation model using customer age and spending amount. The engineer applies min-max normalization to transform each feature into the range 0–1, preventing large numerical scales from dominating distance calculations.

```python
import pandas as pd

data = {
    "Age": [20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)

minimum = df["Age"].min()
maximum = df["Age"].max()

df["Age_Normalized"] = (
    (df["Age"] - minimum) /
    (maximum - minimum)
)

print(df.round(2))
```

We created a DataFrame containing five ages and calculated the minimum and maximum values. We then applied the min-max formula to each age and stored the results in a new column called Age_Normalized. The smallest age became 0, the largest became 1, and the remaining values were mapped proportionally between these limits.

## Model Evaluation

Model Evaluation is the process of measuring how well a machine learning model performs on data that was not used to train it. Statistical concepts help us estimate prediction errors, compare models, measure uncertainty, and determine whether a model is likely to generalize to new data.

#### Numerical Example
Actual values: [100, 200, 300]
Predicted values: [110, 190, 280]
Absolute errors are 10, 10, and 20.
\[
MAE=\frac{10+10+20}{3}=13.33
\]

#### Where It Is Used
- Comparing prediction models.
- Measuring regression errors.
- Evaluating classification performance.
- Checking whether a model generalizes to unseen data.
#### AI/ML Engineer Example
An AI/ML engineer builds two house-price prediction models and compares their MAE on a held-out test dataset. The model with the lower MAE makes smaller average absolute prediction errors on that test set.

```python
actual = [100, 200, 300]
predicted = [110, 190, 280]

absolute_errors = [
    abs(a - p)
    for a, p in zip(actual, predicted)
]

mae = sum(absolute_errors) / len(actual)

print("Absolute errors:", absolute_errors)
print("MAE:", round(mae, 2))
```

We created two lists containing actual values and model predictions. We used zip() to compare corresponding values and calculated each absolute error using abs(). Then, we added the errors and divided by the number of observations to calculate MAE. A lower MAE generally indicates better prediction accuracy when comparing models on the same target and test dataset.

## Linear Regression

Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value by modeling the relationship between input features and a target variable. Statistics provides the least-squares method for estimating the coefficients and tools for examining residuals, uncertainty, and the strength of relationships.

#### Numerical Example
Suppose the fitted relationship between study hours and exam scores is:
\[
\hat{y}=40+10x
\]

For a student who studies 4 hours:
\[
\hat{y}=40+10(4)=80
\]

The predicted score is 80.
#### Where It Is Used
- Predicting house prices.
- Forecasting sales and revenue.
- Estimating salary based on experience.
- Understanding relationships between numerical variables.
#### AI/ML Engineer Example
An AI/ML engineer develops a house-price prediction model using house area as an input feature. The model learns a line that minimizes the sum of squared residuals in the training data and uses that line to predict prices for new houses.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([50, 60, 70, 80, 90])

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("Predicted score:", prediction[0])
```

We initialized LinearRegression() and trained the model using fit(), which estimated the intercept and slope from the sample data. ## Logistic Regression

Logistic Regression is a supervised machine learning algorithm used primarily for classification, especially when predicting one of two outcomes. It uses the logistic function to convert a linear combination of input features into a probability between 0 and 1, which can then be used to assign a class.

#### Numerical Example
Suppose a trained model estimates that a customer has a purchase probability of 0.80.
Using a classification threshold of 0.50:
- Probability = 0.80
- Threshold = 0.50
- Predicted class = 1 (purchase)
#### Where It Is Used
- Predicting customer churn.
- Detecting spam emails.
- Predicting loan default risk.
- Classifying whether a customer will purchase a product.
#### AI/ML Engineer Example
An AI/ML engineer develops a customer-churn model using monthly charges and customer tenure. Logistic regression estimates the probability that each customer will leave, helping the company identify customers who may need additional support.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()

model.fit(X, y)

probability = model.predict_proba([[5]])[0, 1]
prediction = model.predict([[5]])[0]

print("Purchase probability:", round(probability, 3))
print("Predicted class:", prediction)
```

We created an input array containing six numerical observations and a binary target, where 0 represents no purchase and 1 represents a purchase. We trained a logistic regression model using fit(). Then, predict_proba() estimated the probability of class 1 for an input value of 5, while predict() returned the predicted class using the model's default decision threshold. The model learns its coefficients from the data, so the probability should be read from the actual execution output.

## Naive Bayes
Naive Bayes is a probabilistic machine learning algorithm based on Bayes' theorem that predicts a class using the probability of observing the input features under each class. It assumes that the features are conditionally independent given the class, which simplifies probability calculations and makes the algorithm efficient for many classification tasks.
#### Formula
Bayes' theorem:
\[
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}
\]

Where:
- \(P(A\mid B)\) = probability of class \(A\) given evidence \(B\)
- \(P(B\mid A)\) = probability of observing the evidence given class \(A\)
- \(P(A)\) = prior probability of class \(A\)
- \(P(B)\) = probability of the evidence
#### Numerical Example
Suppose 40% of emails are spam. An email contains a word that appears in:
- 80% of spam emails
- 10% of non-spam emails
Using Bayes' theorem:
\[
P(\text{Spam}\mid\text{Word})
=\frac{0.8(0.4)}{0.8(0.4)+0.1(0.6)}
\]

\[
=\frac{0.32}{0.38}\approx0.842
\]

The probability of spam given this word is approximately 84.2% under these assumptions.
#### Where It Is Used
- Spam email classification.
- Text categorization and sentiment analysis.
- Document classification.
- Some medical and other probabilistic classification tasks.
#### AI/ML Engineer Example
An AI/ML engineer develops a spam detector using word frequencies from emails. The Naive Bayes model estimates how likely each word is under the spam and non-spam classes and combines that evidence to classify incoming emails.

```python
from sklearn.naive_bayes import MultinomialNB

# Each column represents a word count:
# [free, meeting]
X = [
    [3, 0],
    [2, 0],
    [0, 3],
    [0, 2]
]

# 1 = spam, 0 = not spam
y = [1, 1, 0, 0]

model = MultinomialNB()

model.fit(X, y)

prediction = model.predict([[2, 0]])

print("Predicted class:", prediction[0])
```

We represented four training emails using two features: the number of times the word free appears and the number of times meeting appears. We assigned class labels to identify spam and non-spam emails and trained a MultinomialNB model. We then passed a new email with two occurrences of free and no occurrences of meeting to predict(). The model predicts class 1, representing spam, based on the patterns learned from the training examples.

## Clustering
Clustering is an unsupervised machine learning technique used to group similar data points together without requiring predefined target labels. Statistical concepts such as distance, averages, variance, and probability help determine how similar observations are and how well they fit into different clusters.
#### Formula
Euclidean distance between two points:
\[
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
\]

#### Numerical Example
Suppose two customers are represented by their monthly spending:
- Customer A: ₹1,000
- Customer B: ₹1,200
- Customer C: ₹10,000
Customers A and B are much closer to each other in spending than either is to Customer C. A clustering algorithm may group A and B together and place C in a different cluster, depending on the data and algorithm.
#### Where It Is Used
- Customer segmentation.
- Grouping similar products.
- Identifying patterns in user behavior.
- Discovering groups in unlabeled datasets.
#### AI/ML Engineer Example
An AI/ML engineer develops a customer-segmentation model using annual income and spending scores. The engineer uses K-Means clustering to group customers with similar characteristics so the business can design more targeted marketing campaigns.

```python
import numpy as np
from sklearn.cluster import KMeans

X = np.array([
    [1000, 10],
    [1200, 12],
    [1100, 11],
    [8000, 80],
    [8500, 85],
    [9000, 90]
])

model = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

model.fit(X)

print("Cluster labels:", model.labels_)
print("Cluster centers:")
print(model.cluster_centers_)
```

We created six customers represented by two features: monthly spending and a spending score. We initialized KMeans with n_clusters=2 to divide the customers into two groups and used fit() to learn the cluster assignments and centers. The labels_ attribute shows the cluster assigned to each customer, while cluster_centers_ displays the average feature values for each cluster. Since spending values have different scales, scaling the features before clustering would generally be advisable in a real project.

## Principal Component Analysis (PCA)

Principal Component Analysis (PCA) is a statistical dimensionality-reduction technique that transforms correlated numerical features into a smaller set of new variables called principal components. These components are constructed to capture as much of the data's variance as possible, allowing us to simplify high-dimensional datasets while retaining important patterns.
#### Formula
The first principal component is a linear combination of the original features:
\[
PC_1=w_1X_1+w_2X_2+\cdots+w_pX_p
\]

The weights are chosen so that the first component captures the maximum possible variance, subject to the component's normalization constraints.
#### Numerical Example
Suppose a dataset contains height in centimeters and height in meters. Both variables contain nearly identical information.
PCA can represent this highly correlated information using one principal component instead of keeping both original features.
#### Where It Is Used
- Reducing the number of numerical features.
- Visualizing high-dimensional data in two or three dimensions.
- Reducing redundancy among correlated features.
- Supporting preprocessing for some machine learning models.
#### AI/ML Engineer Example
An AI/ML engineer works with a dataset containing hundreds of correlated sensor measurements. The engineer applies PCA to reduce the number of dimensions before training a model, potentially reducing computational cost while preserving much of the variation in the original data.

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = np.array([
    [150, 50],
    [160, 60],
    [170, 70],
    [180, 80],
    [190, 90]
])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=1)

X_reduced = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("Reduced shape:", X_reduced.shape)
print("Explained variance ratio:")
print(pca.explained_variance_ratio_)
```

We created a dataset containing five observations and two numerical features. We standardized the features using StandardScaler so that differences in scale would not dominate PCA. We then initialized PCA(n_components=1) and fitted it to the standardized data, transforming the original two columns into one principal component. The output shows that the dataset's shape changed from five rows and two columns to five rows and one column. In this example, the features are perfectly linearly related, so the first principal component captures all the variance.

## Recommendation Systems

A Recommendation System predicts which products, movies, songs, or other items a user may prefer based on information such as previous ratings, purchase history, item similarities, or user behavior. Statistics helps these systems estimate preferences, measure similarities, model uncertainty, and rank items according to their predicted relevance.
#### Formula
A simple user-average rating baseline is:
\[
\hat{r}_{u}=\frac{1}{n_u}\sum_{i=1}^{n_u}r_{ui}
\]

Where:
- \(\hat{r}_{u}\) = estimated average rating for user \(u\)
- \(r_{ui}\) = a rating given by the user
- \(n_u\) = number of ratings provided by the user
This baseline can be extended with item averages, similarities, and collaborative filtering.
#### Numerical Example
Suppose a user has rated three movies:
[4, 5, 3]
Their average rating is:
\[
\hat{r}=\frac{4+5+3}{3}=4
\]

A rating of 4 provides a simple baseline for that user's preferences, although it does not by itself identify which movie they will prefer next.
#### Where It Is Used
- Movie and music recommendations.
- E-commerce product recommendations.
- Personalized news feeds.
- Suggesting content based on user similarities and past interactions.
#### AI/ML Engineer Example
An AI/ML engineer builds a movie recommendation system using user ratings. The engineer analyzes average ratings and similarities between users or movies to recommend titles that a user is likely to enjoy.

```python
import pandas as pd

ratings = pd.DataFrame({
    "User": ["A", "A", "A", "B", "B", "C"],
    "Movie": [
        "Movie1", "Movie2", "Movie3",
        "Movie1", "Movie3", "Movie2"
    ],
    "Rating": [4, 5, 3, 5, 4, 4]
})

user_average = ratings.groupby("User")["Rating"].mean()

print("Average rating by user:")
print(user_average)

movie_average = ratings.groupby("Movie")["Rating"].mean()

print("\nAverage rating by movie:")
print(movie_average.sort_values(ascending=False))
```

We created a ratings DataFrame containing user IDs, movie names, and ratings. We used groupby("User")["Rating"].mean() to calculate each user's average rating and groupby("Movie")["Rating"].mean() to calculate each movie's average rating. We then sorted the movie averages from highest to lowest. This provides a basic statistical summary that can help identify popular movies, but a real personalized recommendation system would also consider which movies each user has already watched and their individual preferences.
