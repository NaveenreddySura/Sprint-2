# Correlation & Covariance

## Correlation

Correlation measures the strength and direction of the relationship between two variables. Its value usually ranges from -1 to +1.

A positive value means both variables tend to increase together, while a negative value means one tends to increase when the other decreases.

### Formula
For Pearson correlation:

r = Cov(X, Y) / (σX × σY)

Where:

r = correlation coefficient
Cov(X,Y) = covariance between X and Y
σX = standard deviation of X
σY = standard deviation of Y


### Numerical Example

Suppose:

Study Hours:  1   2   3   4   5
Marks:       50  55  65  70  80

As study hours increase, marks also increase.

The correlation will be close to:

$$ r \approx +1 $$

indicating a strong positive relationship.

### AI/ML Engineer Example

An ML engineer can check the correlation between house size and house price to understand whether house size may be a useful feature for predicting price.

```python
import numpy as np
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5]
marks = [50, 55, 65, 70, 80]

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Correlation:", correlation)

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

In the Python code, we store study hours and marks in two lists. np.corrcoef() calculates the correlation between the two variables, and [0, 1] extracts the correlation value from the resulting matrix. The scatter plot then visually shows that the points follow an upward pattern.

## Covariance

Covariance measures how two variables change together. A positive covariance means they generally move in the same direction, while a negative covariance means they generally move in opposite directions.

Unlike correlation, covariance does not have a fixed range and depends on the scale of the variables.

### Formula

Population covariance:

$$ Cov(X,Y)= \frac{\sum (x_i-\mu_x)(y_i-\mu_y)}{N} $$


### Numerical Example

Suppose:

X = [1, 2, 3]
Y = [2, 4, 6]

Mean of X:

$$ \bar{x}=2 $$

Mean of Y:

$$ \bar{y}=4 $$

The variables increase together, so their covariance is positive.

### AI/ML Engineer Example

An ML engineer can use covariance to understand whether two features, such as advertising spend and sales, tend to increase or decrease together.

```python
import numpy as np

x = [1, 2, 3]
y = [2, 4, 6]

covariance = np.cov(x, y)[0, 1]

print("Covariance:", covariance)
```

In the Python code, x and y contain the two variables. np.cov() calculates the covariance matrix for them, and [0, 1] extracts the covariance between X and Y. The result of 2.0 indicates that both variables move in the same direction.

## Positive Correlation

Positive Correlation occurs when two variables generally move in the same direction. When one variable increases, the other tends to increase as well.

The correlation coefficient is between:
0<r≤1

### Formula
$$ 0 < r \leq 1 $$

A value close to +1 indicates a strong positive relationship.

### Numerical Example
Advertising Spend:  10  20  30  40  50
Sales:             100 150 200 260 310

As advertising spend increases, sales also increase.

Therefore:

r>0

### AI/ML Engineer Example

An ML engineer may find a positive correlation between house size and house price, which suggests that larger houses tend to have higher prices.

```python
import numpy as np
import matplotlib.pyplot as plt

advertising = [10, 20, 30, 40, 50]
sales = [100, 150, 200, 260, 310]

correlation = np.corrcoef(advertising, sales)[0, 1]

print("Correlation:", correlation)

plt.scatter(advertising, sales)

plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.title("Positive Correlation")

plt.show()
```

The Python code stores advertising spending and sales values. np.corrcoef() calculates their correlation, while plt.scatter() creates a scatter plot. The upward direction of the points visually represents positive correlation.

## Negative Correlation

Negative Correlation occurs when two variables generally move in opposite directions. When one variable increases, the other tends to decrease.

The correlation coefficient is between:

−1≤r<0

### Formula
$$ -1 \leq r < 0 $$

A value close to -1 indicates a strong negative relationship.

### Numerical Example
Exercise Hours:  1   2   3   4   5
Body Fat (%):   25  23  21  19  17

As exercise increases, body fat decreases.

### AI/ML Engineer Example

An ML engineer might find a negative correlation between product price and number of purchases, meaning purchases tend to decrease as price increases.

```python
import numpy as np
import matplotlib.pyplot as plt

exercise = [1, 2, 3, 4, 5]
body_fat = [25, 23, 21, 19, 17]

correlation = np.corrcoef(exercise, body_fat)[0, 1]

print("Correlation:", correlation)

plt.scatter(exercise, body_fat)

plt.xlabel("Exercise Hours")
plt.ylabel("Body Fat (%)")
plt.title("Negative Correlation")

plt.show()
```

The Python code stores exercise hours and body-fat percentages. np.corrcoef() calculates their correlation, which is -1 because the example has a perfectly decreasing linear relationship. The scatter plot shows the points moving downward from left to right.

## No Correlation

No Correlation means there is little or no linear relationship between two variables. Knowing one variable does not help us predict the other through a linear relationship.

### Formula
r≈0

### Numerical Example

Suppose:

Shoe Size:       7   9   6   10   8
Test Score:     75  82  91   68  77

There is no clear linear pattern between shoe size and test score.

Therefore:

r≈0

### AI/ML Engineer Example

An ML engineer may discover that customer ID and purchase amount have almost no meaningful relationship, so customer ID should generally not be treated as a predictive feature.

```python
import numpy as np
import matplotlib.pyplot as plt

shoe_size = [7, 9, 6, 10, 8]
test_score = [75, 82, 91, 68, 77]

correlation = np.corrcoef(shoe_size, test_score)[0, 1]

print("Correlation:", correlation)

plt.scatter(shoe_size, test_score)

plt.xlabel("Shoe Size")
plt.ylabel("Test Score")
plt.title("No Clear Correlation")

plt.show()
```

The Python code calculates the Pearson correlation between shoe size and test score. The scatter plot displays the observations. If the points do not form a clear upward or downward pattern, the correlation is close to zero.

## Pearson Correlation

Pearson Correlation measures the strength and direction of the linear relationship between two numerical variables.

### Formula
r = Cov(X,Y) / (σX × σY)

Where:

r = Pearson correlation coefficient
Cov(X,Y) = Covariance between X and Y
σX = Standard deviation of X
σY = Standard deviation of Y


### Numerical Example
Study Hours = [1, 2, 3, 4, 5]

Marks = [20, 40, 60, 80, 100]

The values increase together in a linear pattern.

Therefore, Pearson correlation will be close to +1.

```python
from scipy.stats import pearsonr

study_hours = [1, 2, 3, 4, 5]
marks = [20, 40, 60, 80, 100]

correlation, p_value = pearsonr(study_hours, marks)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

## Spearman Correlation
Definition
Spearman correlation measures the strength and direction of a relationship using the ranks of the data.

It is useful when:

Data is ordinal.
The relationship is monotonic.
The data may not follow a normal distribution.
We want to work with rankings.

### Formula
ρ = 1 - [6Σd² / n(n² - 1)]

Where:

ρ = Spearman correlation
d = Difference between ranks
n = Number of observations

### AI/ML Engineer Example

An ML engineer can use Spearman correlation to compare feature rankings or analyze relationships where the variables increase/decrease consistently but not necessarily in a straight line.

```python
from scipy.stats import spearmanr

teacher_a = [1, 2, 3, 4, 5]
teacher_b = [1, 2, 3, 5, 4]

correlation, p_value = spearmanr(teacher_a, teacher_b)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

The Python code passes the two rankings to spearmanr(). The function converts the values into ranks and measures how similarly the two rankings change. A value close to 1 means the rankings are highly similar.

## Correlation Matrix
Definition

A Correlation Matrix is a table that shows the correlation between every pair of numerical variables in a dataset.

The diagonal values are always:

$$ 1 $$

because each variable is perfectly correlated with itself.

### Formula

Each cell contains a correlation coefficient:

−1≤r≤1

### Numerical Example

Suppose a dataset contains:

Study Hours
Marks
Sleep Hours
A correlation matrix can show the relationship between every pair of these variables.

### AI/ML Engineer Example

An ML engineer can use a correlation matrix to identify highly correlated features before training a model. For example, annual salary and monthly salary may contain almost the same information.

In the Python code, we first create a DataFrame containing three numerical variables. The df.corr() function calculates the Pearson correlation between every pair of columns. The result is a correlation matrix where each cell represents the relationship between two variables.

## Heatmap

A Heatmap is a graphical representation of numerical values using different shades of color.

In correlation analysis, a heatmap makes it easy to visually identify strong positive, strong negative, and weak correlations.

### Formula

The heatmap itself does not have a separate formula.

It visualizes correlation coefficients:

−1≤r≤1

### AI/ML Engineer Example

An ML engineer can use a correlation heatmap before model training to quickly identify highly correlated features and decide whether some redundant features should be removed.

Exploratory Data Analysis
Feature selection
Finding multicollinearity
Understanding relationships between many variables
