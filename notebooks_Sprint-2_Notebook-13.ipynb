# Statistical Data Visualization 
Statistical data visualization is the process of presenting data in charts and graphs to understand patterns, trends, distributions, and relationships between variables. In Data Science and AI/ML, visualizations help engineers explore datasets, identify outliers, compare categories, and understand data before building machine learning models.

## Histogram
Purpose
A Histogram displays the distribution of numerical data by dividing values into intervals called bins and showing how many observations fall within each interval. It helps us understand the shape, spread, concentration, and possible skewness of a dataset.
## When to Use
- To understand the distribution of numerical data.
- To identify frequently occurring value ranges.
- To check whether data is symmetric or skewed.
- To investigate possible outliers or unusual patterns.
## Advantages
- Easy to understand the distribution of large datasets.
- Helps identify skewness and concentration.
- Useful during Exploratory Data Analysis (EDA).
## Limitations
- The appearance changes depending on the number and width of bins.
- Exact individual values are not shown.
- Overlapping groups can be difficult to compare in one histogram.

```python
import matplotlib.pyplot as plt

exam_scores = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(exam_scores, bins=5, edgecolor="black")
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation
The histogram groups exam scores into five intervals and shows the number of students in each interval. Taller bars indicate score ranges with more students, while shorter bars indicate ranges with fewer students. The overall shape helps us understand whether scores are concentrated in a particular range or spread across the distribution.

## Bar Chart
Purpose
A Bar Chart represents categorical data using rectangular bars, where the height or length of each bar represents a value such as frequency, sales, or revenue. It is useful for comparing different categories and identifying which categories have the highest or lowest values.
### When to Use
- To compare sales across products.
- To display the number of customers in different regions.
- To compare performance across departments.
- To show counts or summary values for categories.
### Advantages
- Simple to read and interpret.
- Makes comparisons between categories easy.
- Works well with a small or moderate number of categories.
### Limitations
- Too many categories can make the chart cluttered.
- It does not show the distribution of individual numerical observations.
- Comparisons can be misleading if axes are designed poorly.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [50, 80, 30, 60]

plt.bar(products, sales)
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.show()
```

### Interpretation
The chart compares the units sold for four products. Mobile has the highest sales at 80 units, while Tablet has the lowest at 30 units. This helps a business identify its best-selling and lowest-selling products.

## Line Chart
Purpose
A Line Chart displays data points connected by lines to show how a numerical value changes across an ordered sequence, usually time. It is commonly used to identify trends, growth, decline, and fluctuations in business or operational data.
### When to Use
- To track monthly or yearly sales.
- To monitor website traffic over time.
- To analyze stock prices or production trends.
- To compare changes in performance across time periods.
### Advantages
- Makes trends easy to identify.
- Shows increases, decreases, and fluctuations clearly.
- Allows multiple time series to be compared.
### Limitations
- Too many lines can make the chart difficult to read.
- Missing values can complicate interpretation.
- Connecting unordered categories with lines can suggest a trend that does not exist.

#### Interpretation
The line chart shows monthly revenue increasing overall from 10,000 in January to 18,000 in May. Revenue decreases slightly in March before rising again. The business can use this chart to understand revenue trends and identify periods of decline or growth.

## Scatter Plot
Purpose
A Scatter Plot displays the relationship between two numerical variables using individual points. Each point represents one observation, helping us identify positive or negative relationships, clusters, unusual observations, and possible nonlinear patterns.
#### When to Use
- To study the relationship between advertising spending and sales.
- To compare study hours and exam scores.
- To examine house area and house price.
#### Advantages
- Makes relationships between two numerical variables visible.
- Helps identify clusters and potential outliers.
- Can reveal patterns that summary statistics may hide.
#### Limitations
- Large datasets may produce overlapping points.
- Correlation does not prove causation.
- Relationships involving more than two variables may require additional visual encoding.

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
scores = [45, 50, 60, 65, 75, 85]

plt.scatter(hours, scores)

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

#### Interpretation
Each point represents a student's study hours and exam score. The points generally move upward as study hours increase, indicating a positive relationship between the two variables. This suggests that students who study more tend to score higher in this sample, but it does not establish that study hours alone cause higher scores.

## Pie Chart

#### Purpose
A Pie Chart represents how a total is divided among different categories using slices of a circle. Each slice represents a category's proportion or percentage of the whole.
#### When to Use
- To show the percentage of sales contributed by different products.
- To display a company's budget allocation.
- To compare the market share of a small number of categories.
#### Advantages
- Simple to understand when there are only a few categories.
- Shows each category's proportion of the total.
- Useful for presenting percentage-based summaries.
#### Limitations
- Difficult to compare slices with similar sizes.
- Too many categories make the chart cluttered.
- Not ideal for showing changes over time or precise comparisons.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [50, 80, 30, 40]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Sales Share")
plt.show()
```

#### Interpretation
The pie chart shows each product's percentage of total sales. Phones contribute the largest share, approximately 40%, while tablets contribute 15%. The company can use this information to understand how total sales are distributed across products.

## Density Plot
Purpose
A Density Plot displays a smoothed estimate of the distribution of numerical data. Unlike a histogram, which uses bins, a density plot uses a curve to show where observations are concentrated and how the distribution is shaped.
#### When to Use
- To examine the distribution of salaries or house prices.
- To compare distributions between customer groups.
- To identify skewness and multiple peaks in numerical data.
#### Advantages
- Provides a smooth view of a distribution.
- Makes it easier to compare distributions using overlapping curves.
- Helps identify the location of peaks and the spread of values.
#### Limitations
- The curve depends on the smoothing bandwidth.
- Small datasets can produce misleading shapes.
- Density is not the same as the number of observations; the total area under a properly normalized density curve equals 1.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [50, 80, 30, 40]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Sales Share")
plt.show()
```

#### Interpretation
The pie chart shows each product's percentage of total sales. Phones contribute the largest share, approximately 40%, while tablets contribute 15%. The company can use this information to understand how total sales are distributed across products.

## Box Plot
Purpose
A Box Plot summarizes the distribution of numerical data using the median, first quartile, third quartile, and whiskers. It helps us understand the spread of values and identify potential outliers without displaying every individual observation.
#### When to Use
- To compare salaries across departments.
- To identify potential outliers in a dataset.
- To compare the spread of exam scores across classes.
#### Advantages
- Summarizes a distribution in a compact chart.
- Makes medians and interquartile ranges easy to compare.
- Helps identify potential outliers and differences in spread.
#### Limitations
- Does not show every observation.
- Can hide clusters or multiple peaks.
- Points beyond the whiskers are potential outliers, not automatically errors.

```python
import matplotlib.pyplot as plt

salaries = [25000, 28000, 30000, 32000, 35000,
            36000, 38000, 40000, 45000, 90000]

plt.boxplot(salaries)

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

#### Interpretation
The box shows the middle 50% of salaries, with the line inside representing the median. The whiskers show the range of non-outlier values according to the box plot's rule, while the value 90,000 may appear as a potential outlier. This helps an analyst investigate unusually high salaries or possible data-entry issues.

## Violin Plot
Purpose
A Violin Plot combines a box-plot-style summary with a smoothed density shape to show the distribution of numerical values. The width of the violin at different values indicates the relative concentration of observations around those values.
#### When to Use
- To compare salary distributions across departments.
- To compare model errors across different algorithms.
- To examine differences in customer spending across segments.
#### Advantages
- Shows distribution shape as well as central tendency.
- Can reveal multiple peaks that a box plot may hide.
- Useful for comparing distributions across several groups.
####  Limitations
- More difficult to interpret for beginners than a box plot.
- The shape depends on density smoothing.
- Can be misleading when sample sizes are very small.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Department": [
        "IT", "IT", "IT", "IT", "IT",
        "HR", "HR", "HR", "HR", "HR"
    ],
    "Salary": [
        40000, 45000, 48000, 50000, 70000,
        30000, 32000, 35000, 36000, 40000
    ]
}

df = pd.DataFrame(data)

sns.violinplot(data=df, x="Department", y="Salary")

plt.title("Salary Distribution by Department")
plt.show()
```

#### Interpretation
The violin plot compares salary distributions in the IT and HR departments. Wider portions of each violin indicate salary ranges with greater estimated density. The IT salaries in this example tend to be higher, but the plot should be interpreted alongside sample size and other relevant business information.

## Pair Plot
Purpose
A Pair Plot displays pairwise relationships between multiple numerical features in a dataset using a grid of scatter plots. The diagonal usually displays each feature's distribution, helping us explore relationships, clusters, and patterns across several variables at once.
#### When to Use
- During exploratory data analysis before training a machine learning model.
- To examine relationships among house area, bedrooms, and house price.
- To identify possible correlations and clusters in a dataset.
#### Advantages
- Shows relationships between many numerical features in one visualization.
- Helps identify patterns and potential outliers.
- Provides both individual feature distributions and pairwise comparisons.
#### Limitations
- Becomes crowded when many features are included.
- Can be slow to generate for large datasets.
- Shows only pairwise relationships and may not reveal complex interactions among several features.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Area": [1000, 1500, 2000, 2500, 3000],
    "Bedrooms": [2, 3, 3, 4, 5],
    "Price": [20, 30, 42, 55, 65]
}

df = pd.DataFrame(data)

sns.pairplot(df)

plt.show()
```

#### Interpretation
The pair plot displays the relationships between area, bedrooms, and price in separate scatter plots, with individual feature distributions along the diagonal. We can examine whether house prices tend to increase with area, whether bedrooms are related to area, and whether any unusual observations appear. This helps an AI/ML engineer understand the dataset before selecting features for a prediction model.

## Heatmap
Purpose
A Heatmap represents numerical values using different colors in a grid, making patterns and differences easier to identify. In statistical analysis, it is often used to visualize a correlation matrix and quickly identify strongly positive or negative relationships between numerical features.
#### When to Use
- To visualize a correlation matrix.
- To identify highly correlated features before building a machine learning model.
- To compare numerical values across categories or groups.
#### Advantages
- Makes patterns in large tables easier to recognize.
- Helps identify strong positive and negative correlations quickly.
- Allows many feature relationships to be displayed in a compact format.
#### Limitations
- Color differences can be misleading if the color scale is poorly chosen.
- Correlation heatmaps show linear relationships, not necessarily all types of relationships.
- Correlation does not establish causation.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Hours_Studied": [1, 2, 3, 4, 5],
    "Exam_Score": [50, 60, 70, 80, 90],
    "Sleep_Hours": [8, 7, 8, 6, 7]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)
```

#### Interpretation
The heatmap displays the correlation coefficient between each pair of features, with the numerical values shown inside the cells. Values near +1 indicate strong positive linear relationships, values near -1 indicate strong negative linear relationships, and values near 0 indicate weak linear relationships. The diagonal cells are always 1 because each feature is perfectly correlated with itself.
