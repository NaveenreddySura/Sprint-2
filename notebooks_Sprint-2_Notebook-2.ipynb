## Mean
Mean is the most commonly used measure of central tendency. It is calculated by adding all the values in a dataset and dividing the total by the number of values.

#formula
Mean=Sum of all values/Number of values

#Number of values
Data:

10, 20, 30, 40, 50

Step 1: Add all values:

10+20+30+40+50=150 

Step 2: Number of values = 5

Mean = 150/5=30 

Answer: Mean = 30

```python
# Python Implementation

numbers = [10, 20, 30, 40, 50]

mean = sum(numbers) / len(numbers)

print("Mean:", mean)
```

## Median
Median is the middle value of a dataset after arranging the values in ascending or descending order. If there are an even number of values, the median is calculated by taking the average of the two middle values.

#Formula

For an odd number of values:

Median = Middle Value

For an even number of values:

Median= (First Middle Value + Second Middle Value)/2


```python
#Python Implementation
import statistics

numbers = [10, 20, 30, 40, 50]

median = statistics.median(numbers)

print("Median:", median)
```

## Mode
Mode is the value that appears most frequently in a dataset. It is especially useful when we want to know the most common or popular value.

```python
#Python Implementation
import statistics

numbers = [10, 20, 20, 30, 40, 20, 50]

mode = statistics.mode(numbers)

print("Mode:", mode)
```

## Weighted Mean
Weighted Mean is an average where different values are given different levels of importance or weights. A value with a higher weight has a greater effect on the final average.

```python
marks = [80, 70, 90]
weights = [0.5, 0.3, 0.2]

weighted_mean = sum(
    mark * weight
    for mark, weight in zip(marks, weights)
) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

## Geometric Mean
Geometric Mean is an average calculated by multiplying all the values together and then taking the nth root, where n is the number of values. It is mainly useful for growth rates, percentages, ratios, and compound changes.

```python
#Formula
import math

growth = [1.10, 1.20]

geometric_mean = math.prod(growth) ** (1 / len(growth))

print("Geometric Mean:", geometric_mean)
print("Growth Rate:", (geometric_mean - 1) * 100, "%")

```

## Harmonic Mean
Harmonic Mean is an average calculated using the reciprocal of each value. It is particularly useful for rates, ratios, speeds, and other values where the denominator is important.

```python
import statistics

speeds = [60, 40]

harmonic_mean = statistics.harmonic_mean(speeds)

print("Harmonic Mean:", harmonic_mean)
```
