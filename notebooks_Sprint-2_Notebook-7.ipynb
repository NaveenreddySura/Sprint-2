# Sampling
Sampling is an important statistical concept because studying an entire population is often expensive, slow, or impossible. Instead, we select a smaller group called a sample and use it to understand the population.

## Sampling

Sampling is the process of selecting a smaller group of observations from a larger population for analysis. The selected group is called a sample, while the complete group is called the population.

### Formula
There is no single formula for sampling.

Sample ⊂ Population



### Numerical Example
Suppose a company has 10,000 customers.

Studying all 10,000 customers may take a lot of time.

The company selects 500 customers for a survey.

Population = 10,000 customers

Sample = 500 customers

```python
import numpy as np

population = np.arange(1, 10001)

sample = np.random.choice(population, size=500, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 10 Sample Values:", sample[:10])
```

### AI/ML Example
Machine learning datasets can contain millions of records.

Sampling can be used to create a smaller dataset for initial analysis and experimentation.

## Random Sampling

Random Sampling is a sampling method where every member of the population has an equal or known chance of being selected. Selection is made randomly to reduce personal bias.

### Formula
There is no specific probability formula required for basic random sampling.

For simple random sampling:

Probability of selection = Sample Size / Population Size

### Numerical Example
Population = 1,000 customers

Sample = 100 customers

Probability of a particular customer being selected:

100 / 1000 = 0.10

Therefore, the probability is 10%.

### AI/ML Engineer Example

Random sampling can be used to select a representative subset of millions of records for exploratory data analysis or model development.

```python
import random

employees = list(range(1, 1001))

sample = random.sample(employees, 100)

print("Number of Selected Employees:", len(sample))
print("Selected Employees:", sample[:10])
```

In the Python code, range(1, 1001) creates a population of 1,000 employees. random.sample() randomly selects 100 different employees from this population without repeating an employee. This gives every employee an opportunity to be included in the sample.

## Stratified Sampling

Stratified Sampling divides the population into meaningful groups called strata and then selects samples from each group. It ensures important groups are represented in the final sample.

### Formula
Sample from each stratum can be selected proportionally:

Stratum Sample Size = (Stratum Population / Total Population) × Total Sample

### Numerical Example
Suppose a company has:

Department A = 600 employees
Department B = 300 employees
Department C = 100 employees

Total = 1,000 employees

If we need a sample of 100 employees:

Department A:

600 / 1000 × 100 = 60

Department B:

300 / 1000 × 100 = 30

Department C:

100 / 1000 × 100 = 10

### AI/ML Example
Stratified sampling is useful when a dataset contains important groups that should all be represented.

For example, when splitting an imbalanced classification dataset, stratification can help preserve class proportions.

```python
departments = {
    "A": 600,
    "B": 300,
    "C": 100
}

total_population = sum(departments.values())
sample_size = 100

for department, population in departments.items():
    sample = (population / total_population) * sample_size
    print(department, "Sample Size:", int(sample))
```

## Cluster Sampling

Cluster Sampling divides a population into naturally occurring groups called clusters and randomly selects some entire clusters for study.

Unlike stratified sampling, we usually study all or many members within the selected clusters.

### Formula

There is no single standard formula for selecting clusters.

A simple selection probability can be expressed as:

P(cluster selected)= Number of Selected Clusters/Total Number of Clusters
	


### Numerical Example

Suppose a company has offices in 10 cities and wants to study employee satisfaction.

Instead of selecting employees from every city, it randomly selects 3 cities:

Bangalore
Hyderabad
Chennai

Then it surveys employees from those selected offices.

$$ P(\text{city selected})=\frac{3}{10}=0.3 $$

So there is a 30% chance for each city to be selected if selection is equally likely.

### AI/ML Engineer Example

Cluster sampling can be useful when working with geographically distributed data, such as selecting certain cities, stores, hospitals, or branches for analysis.

```python
import random

cities = [
    "Bangalore",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Delhi",
    "Pune",
    "Kolkata",
    "Jaipur",
    "Kochi",
    "Ahmedabad"
]

selected_clusters = random.sample(cities, 3)

print("Selected Clusters:", selected_clusters)
```

n the Python code, each city represents a cluster. random.sample() randomly selects three cities from the ten available cities. The selected cities represent the clusters that would be studied.

## Systematic Sampling

Systematic Sampling selects observations at a regular interval from an ordered population. Usually, we choose a random starting point and then select every kth observation.

### Formula
$$ k=\frac{N}{n} $$

where:

N = population size
n = required sample size
k = sampling interval

### Numerical Example

Suppose:

$$ N=1000 $$

and we need:

$$ n=100 $$

Then:

$$ k=\frac{1000}{100}=10 $$

So we select every 10th person after choosing a starting point.

### AI/ML Engineer Example

Systematic sampling can be used to select records from large ordered datasets, such as selecting every 100th transaction for quality checking.

```python
population = list(range(1, 1001))

sample_size = 100

interval = len(population) // sample_size

start = 5

sample = population[start - 1::interval]

print("Sampling Interval:", interval)
print("First 10 Selected:", sample[:10])
```

In the Python code, we create 1,000 population values. The sampling interval is calculated by dividing the population size by the required sample size, giving 10. Starting from customer 5, the slicing operation selects every 10th customer, producing the systematic sample.

## Sampling Bias

Sampling Bias occurs when the selected sample does not properly represent the population because some members have a greater or smaller chance of being selected.

It can lead to misleading conclusions about the entire population.

### Formula

There is no single formula for sampling bias.

Conceptually:

Sampling Bias=Expected Sample Result−True Population Result

### Numerical Example

Suppose a company's actual customer satisfaction is:

$$ 70\% $$

But the company surveys only its most loyal customers and gets:

$$ 90\% $$

The difference is:

$$ 90\%-70\%=20\% $$

This indicates that the sample result is biased upward.

### AI/ML Engineer Example

Sampling bias can cause biased training data, resulting in an ML model that performs poorly for groups that were underrepresented in the training dataset.

```python
population_satisfaction = 70
sample_satisfaction = 90

bias = sample_satisfaction - population_satisfaction

print("Sampling Bias:", bias, "%")
```

In the Python code, we store the true population satisfaction as 70 and the biased sample result as 90. We subtract the population value from the sample value to calculate the difference of 20 percentage points. This illustrates how an unrepresentative sample can produce a misleading result.

## Sampling Error

Sampling Error is the difference between a statistic calculated from a sample and the actual population parameter. It happens naturally because we study only part of the population.

Sampling error is not necessarily caused by a mistake.

### Formula
Sampling Error=Sample Statistic−Population Parameter

### Numerical Example

Suppose the actual average customer spending is:

$$ ₹2,000 $$

A sample of customers gives an average of:

$$ ₹1,950 $$

Therefore:

$$ \text{Sampling Error}=1950-2000 $$ $$ =-₹50 $$

The magnitude of the error is ₹50.

### AI/ML Engineer Example

Sampling error is important when an ML engineer uses a sample to estimate characteristics of a much larger dataset.

```python
population_mean = 2000
sample_mean = 1950

sampling_error = sample_mean - population_mean

print("Sampling Error:", sampling_error)
print("Absolute Error:", abs(sampling_error))
```

In the Python code, we store the population mean and sample mean. We subtract the population mean from the sample mean to calculate the sampling error. The result is -50, meaning the sample estimate is ₹50 lower than the actual population value. abs() gives the magnitude of the error as ₹50.

## Sample Size

Sample Size is the number of observations or individuals included in a sample. Choosing an appropriate sample size helps balance accuracy, cost, time, and computational resources.

### Formula

For a large population, a commonly used sample-size formula for estimating a proportion is:

$$ n=\frac{Z^2p(1-p)}{E^2} $$

where:

n = sample size
Z = Z-score for confidence level
p = estimated proportion
E = margin of error

### For example, for 95% confidence:

$$ Z=1.96 $$

If we assume:

$$ p=0.5 $$

and want:

$$ E=0.05 $$

then:

$$ n=\frac{1.96^2(0.5)(0.5)}{0.05^2} $$ $$ n\approx384.16 $$

So we need approximately:

$$ \boxed{385} $$

observations.

### AI/ML Engineer Example

Sample size helps an ML engineer determine how much data is needed for reliable analysis or model training while avoiding unnecessary computational cost

```python
import math

Z = 1.96
p = 0.5
E = 0.05

sample_size = (Z**2 * p * (1 - p)) / (E**2)

print("Required Sample Size:", math.ceil(sample_size))
```

In the Python code, Z = 1.96 represents a 95% confidence level, p = 0.5 is used as a conservative assumption when the population proportion is unknown, and E = 0.05 represents a 5% margin of error. The formula calculates the required sample size as approximately 384.16, and math.ceil() rounds it up to 385 because we cannot survey a fraction of a person.
