# Feature Relationships 
Feature relationships help us understand how different input variables are connected to each other and how they influence a machine learning model's predictions. These concepts are useful for identifying redundant features, reducing unnecessary data, improving model performance, and making models easier to understand.

## Correlation Matrix

A Correlation Matrix is a table that shows the strength and direction of the relationship between every pair of numerical features in a dataset. Its values usually range from -1 to +1, helping us identify positively related, negatively related, and unrelated features.

### Formula
Pearson correlation coefficient:
\[
r=\frac{\operatorname{Cov}(X,Y)}{\sigma_X\sigma_Y}
\]

Where:
- \(r\) = correlation coefficient
- \(\operatorname{Cov}(X,Y)\) = covariance between two features
- \(\sigma_X,\sigma_Y\) = standard deviations of the features

### Where It Is Used
- Identifying relationships between numerical features.
- Detecting potentially redundant features.
- Performing exploratory data analysis (EDA).
- Understanding patterns before training a machine learning model.

### AI/ML Engineer Example
An AI/ML engineer building a house-price prediction model checks the correlation between house area, number of bedrooms, number of bathrooms, and house price. This helps the engineer understand which features move together and which features may provide overlapping information.

```python
import pandas as pd

data = {
    "Hours_Studied": [1, 2, 3, 4, 5],
    "Exam_Score": [50, 60, 70, 80, 90],
    "Sleep_Hours": [8, 7, 8, 6, 7]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
```

## Multicollinearity

Multicollinearity occurs when two or more independent features in a dataset are highly correlated with one another. It can make it difficult for certain machine learning models, especially linear and logistic regression, to estimate the individual contribution of each feature reliably.

### Formula
There is no single formula for multicollinearity, but it is commonly assessed using correlation coefficients and Variance Inflation Factor (VIF).
For two features, a high absolute correlation such as \(|r| > 0.8\) can be a warning sign, although the appropriate threshold depends on the dataset and application.

### Numerical Example
Suppose a house-price dataset contains these features:
House	Area (sq. ft.)	Area (sq. m.)
A	1,000	92.9
B	1,500	139.4
C	2,000	185.8
Both features represent the same underlying measurement in different units. Their correlation is approximately +1, so including both may introduce redundant information into a model.

### Where It Is Used
- Checking relationships among input features.
- Improving the stability of regression models.
- Identifying redundant columns in datasets.
- Supporting feature selection and model interpretation.

### AI/ML Engineer Example
An AI/ML engineer building a house-price prediction model finds that Area_SqFt and Area_SqM contain almost identical information. The engineer investigates multicollinearity and may remove one feature to make the regression model easier to interpret.

```python
import pandas as pd

data = {
    "Area_SqFt": [1000, 1500, 2000, 2500, 3000],
    "Area_SqM": [92.9, 139.4, 185.8, 232.3, 278.7],
    "Bedrooms": [2, 3, 3, 4, 5]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix.round(2))

if abs(correlation_matrix.loc["Area_SqFt", "Area_SqM"]) > 0.8:
    print("High correlation: possible multicollinearity")
```

We created a DataFrame containing house area in square feet, house area in square meters, and the number of bedrooms. We used df.corr() to calculate the correlation matrix and .round(2) to display the values to two decimal places. Then, we checked whether the correlation between the two area features was greater than 0.8 in absolute value. Since both features measure almost the same thing, the code prints a warning about possible multicollinearity. This is a warning for further investigation, not proof that a feature must always be removed.

## Variance Inflation Factor (VIF)
Definition
Variance Inflation Factor (VIF) is a statistical measure used to determine how much the variance of a regression coefficient is increased because of relationships between a feature and the other input features. A high VIF indicates that a feature may be explained substantially by other features, making its individual regression coefficient harder to estimate reliably.

### Formula
VIF = 1 / (1 - R²)

### Business Example
If Revenue and Monthly Sales have a very high VIF, they may contain similar information.

### AI/ML Example
VIF can be used before Linear Regression to identify features that are highly related to other input features.

### Interpretation
VIF helps us decide whether some features should be removed from a model.

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

data = {
    "Area": [1000, 1500, 2000, 2500, 3000, 3500],
    "Bedrooms": [2, 3, 3, 4, 5, 5],
    "Bathrooms": [1, 2, 2, 3, 3, 4]
}

df = pd.DataFrame(data)

vif_data = pd.DataFrame()
vif_data["Feature"] = df.columns

vif_data["VIF"] = [
    variance_inflation_factor(df.values, i)
    for i in range(df.shape[1])
]

print(vif_data)
```

Finally, we printed the resulting table so that we can identify features that may have high multicollinearity.

## Feature Importance

Feature Importance is a technique used to identify which input features contribute most to a machine learning model's predictions. It helps engineers understand which variables are useful for prediction and which features may have relatively little influence on the model's performance.

### Example
For predicting house prices:

House Size
Number of Bedrooms
Location
Age of House
House Size may be more important than the number of bedrooms.

### Where It Is Used
- Understanding which features a model relies on.
- Explaining model predictions.
- Identifying potentially less useful features.
- Supporting feature selection and model optimization.

### AI/ML Engineer Example
An AI/ML engineer develops a house-price prediction model using house area, location score, number of bedrooms, and property age. After training a Random Forest model, the engineer examines feature importance to understand which variables the model uses most heavily for its predictions.

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

data = {
    "Area": [1000, 1500, 2000, 2500, 3000, 3500],
    "Bedrooms": [2, 3, 3, 4, 5, 5],
    "Location_Score": [5, 6, 7, 8, 8, 9]
}

df = pd.DataFrame(data)

X = df[["Area", "Bedrooms", "Location_Score"]]
y = [20, 30, 42, 55, 65, 80]

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)
```

We first created a DataFrame containing house area, number of bedrooms, and location score, with house prices stored separately as the target variable y. We then used RandomForestRegressor() to create a Random Forest model and trained it using model.fit(X, y). After training, model.feature_importances_ returned an importance score for each input feature. We stored the scores in a DataFrame, sorted them from highest to lowest, and printed the result to identify which features the model relied on most according to its built-in importance measure.

### Feature Selection Basics

Feature Selection is the process of choosing the most useful input features from a dataset while removing irrelevant, redundant, or less informative features. It can reduce model complexity, improve training efficiency, and sometimes improve generalization by reducing unnecessary information.

### Example
Suppose we have 20 features but only 5 are useful for prediction.

We can select the 5 useful features and remove the remaining features.

### Business Example
For customer churn prediction, useful features may include:

Customer Age
Monthly Charges
Contract Type
Customer Tenure
Unnecessary features can be removed.

### AI/ML Example
Feature selection is commonly used before training ML models.

### Interpretation
Good feature selection helps the model focus on the most useful information.

### Where It Is Used
- Removing irrelevant or redundant input columns.
- Reducing training time and memory usage.
- Simplifying machine learning models.
- Reducing overfitting risk when unnecessary features introduce noise.

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression

data = {
    "Area": [1000, 1500, 2000, 2500, 3000, 3500],
    "Bedrooms": [2, 3, 3, 4, 5, 5],
    "Location_Score": [5, 6, 7, 8, 8, 9],
    "House_ID": [101, 102, 103, 104, 105, 106]
}

df = pd.DataFrame(data)

X = df[["Area", "Bedrooms", "Location_Score", "House_ID"]]
y = [20, 30, 42, 55, 65, 80]

selector = SelectKBest(
    score_func=f_regression,
    k=3
)

X_selected = selector.fit_transform(X, y)
```

We created a DataFrame containing four features: area, bedrooms, location score, and house ID. We separated the input features into X and the house prices into the target variable y. We then used SelectKBest() with f_regression to score each feature based on its univariate linear relationship with the target and selected the top three features using k=3. The method fit_transform() returned the selected feature values, while get_support() identified which original columns were selected. Finally, we printed the selected feature names and transformed data.
